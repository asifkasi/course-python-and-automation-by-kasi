```python
import requests
from bs4 import BeautifulSoup
headers={
    "User-agent":"Orthogaffe"
}
response=requests.get("https://en.wikipedia.org/wiki/Python",headers=headers)
response.text
soup=BeautifulSoup(response.text,"html.parser")
for h2 in soup.find_all("h2"):
    var=h2.get_text(strip=True)
    print(var)
```
