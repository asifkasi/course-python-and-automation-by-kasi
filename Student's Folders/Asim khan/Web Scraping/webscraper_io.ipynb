## 4.1 Extract first Table of a page & save into csv file.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
r=requests.get('https://webscraper.io/test-sites/tables')
r.status_code
```

```python
soup=BeautifulSoup(r.text,'html.parser')
web=soup.find('table',class_='table table-bordered')
rows=web.find('tbody').find_all('tr')
list=[]
for tr in rows:
    col=tr.find_all('td')
    dic={
        '#':col[0].text.strip(),
        'Firsname':col[1].text.strip(),
        'Lastname':col[2].text.strip(),
        'Username':col[3].text.strip(),
    }
    list.append(dic)
print(list)
```

```python
df=pd.DataFrame(list)
df
```

```python
df.to_csv('webscraperd_q1.csv',index=False)
```

## a5.2 Extract product details from first row & save into csv file.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
main_url='https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops'
r=requests.get(main_url)
soup=BeautifulSoup(r.text,'html.parser')
products=soup.find_all('div',class_='col-md-4 col-xl-4 col-lg-4')
list=[]
for i in products[0:3]:
    var_dic={
        'product_name':'',
        'product_para':'',
        'product_img':'',
        'product_url':'',
        'product_price':'',
        'procut_review':'',
}
    var_dic['product_name']=i.find('a').text.strip()
    var_dic['product_para']=i.find('p',class_='description card-text').text.strip()
    var_dic['product_img']=i.find('img')['src']
    var_dic['product_url']=i.find('a')['href']
    url='https://webscraper.io'
    actual_url=url+var_dic['product_url']
    web=requests.get(actual_url)
    new_soup1=BeautifulSoup(web.text,'html.parser')
    var_dic['product_name']=new_soup1.find('h4',itemprop='name').text.strip()
    var_dic['product_para']=i.find('p',class_='description card-text').text.strip()
    var_dic['product_img']=i.find('img')['src']
    var_dic['product_price']=i.find('span').text.strip()
    var_dic['procut_review']=i.find('p',class_='review-count float-end').text.strip()
    list.append(var_dic)
list

    

```
