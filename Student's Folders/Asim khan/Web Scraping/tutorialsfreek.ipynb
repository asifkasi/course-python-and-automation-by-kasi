```python
import requests
from bs4 import BeautifulSoup
web=requests.get("https://www.tutorialsfreak.com/")
web.status_code
```

```python
soup=BeautifulSoup(web.text,"html.parser")
print(soup.prettify())
```

```python
lines=soup.find_all('p')
for i in lines:
    print(i.text)
```

```python
var=soup.find("div",class_="wrapper")
print(var.text)
```

```python
var1=soup.find('section',class_="why-choose-section")
var1.text
```

```python
headings=var1.find_all('p')
headings
```

```python
for i in headings:
    print(i.text)
```

```python
handp=soup.find('div',class_='order-1 order-lg-0 col-lg-8 col-md-12')
# Agar div mil jaye
if handp:
    # Saare h3 headings ke liye loop
    for heading in handp.find_all('h3'):
        print(heading.get_text(strip=True))

        # heading ke baad jo next tag ho, wo paragraph ho sakta hai
        next_p = heading.find_next_sibling('p')
        if next_p:
            print(next_p.get_text(strip=True))
        print("\n")
else:
    print("Content section not found!")
```
