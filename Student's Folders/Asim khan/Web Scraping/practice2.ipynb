```python
import requests
from bs4 import BeautifulSoup
header={
    "User-agent": "Orthogaffe"
}
r=requests.get("https://en.wikipedia.org/wiki/Python",headers=header)
r
soup=BeautifulSoup(r.text,"html.parser")
for i in soup.find_all('h2'):
    var=i.get_text(strip=True)
    print(var)
```

```python
ul_par=soup.find('div',class_="mw-content-ltr mw-parser-output")
ul_par
```

```python
uls=ul_par.find_all('ul')
len(uls)
```

```python
for i in uls:
    print(i.text)
```
