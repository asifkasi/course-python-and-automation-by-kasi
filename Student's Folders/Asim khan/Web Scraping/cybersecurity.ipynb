```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
r=requests.get('https://www.cisa.gov/news-events/cybersecurity-advisories')
r.status_code
```

```python
soup=BeautifulSoup(r.text,'html.parser')
web=soup.find_all('div',class_='c-view__row')
data=[]
for div in web:
    dic={
        "date":"",
        "name":"",
        "url":"",
    }
    dic['date']=div.find('time').text
    dic['name']=div.find('span').text
    dic['url']=div.find('a')['href']
    data.append(dic)
    #print(dic)
```

```python
df=pd.DataFrame(data)
df
```

```python
df.to_csv('cybersecurity.csv',index=False)
```

```python
d=requests.get('https://www.cisa.gov/news-events/cybersecurity-advisories?page=1')
soup=BeautifulSoup(d.text,'html.parser')
page1=soup.find_all('div',class_='c-view__row')
data1=[]
for div in page1:
    dic1={
        "date":"",
        "name":"",
        "url":"",
    }
    dic1['date']=div.find('time').text
    dic1['name']=div.find('span').text
    dic1['url']=div.find('a')['href']
    data1.append(dic1)
    print(data1)
```

```python
final_dat=data+data1 #merging two list into one
df1=pd.DataFrame(final_dat)
df1
```

```python
df1.to_csv('cybersecurity.csv',index=False)
```
