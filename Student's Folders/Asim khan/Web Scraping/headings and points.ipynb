```python
import requests
from bs4 import BeautifulSoup
header={
    "User-Agent": "IsraBot"
}
r=requests.get("https://en.wikipedia.org/wiki/Python",headers=header)
soup=BeautifulSoup(r.text,'html.parser')
sub_headings=soup.find_all('div',class_='mw-heading mw-heading2')
for i in sub_headings:
    var=i.find('h2')
    #print(var)
    if var:
        print(var.text+":")
        ul=i.find_next_sibling('ul')
        if ul:
            for li in ul.find_all('li'):
                text = li.get_text(strip=True)
                link_tag = li.find('a')
                if link_tag and link_tag.has_attr('href'):
                    link = link_tag['href']
                    # Wikipedia links often start with '/wiki/', so add base URL
                    if link.startswith('/wiki/'):
                        link = "https://en.wikipedia.org" + link
                    print(f"  - {text} ({link})")
                else:
                    print(f"  - {text}")
```
