```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time
from selenium import webdriver

driver = webdriver.Chrome()

url   = "https://search.cpsa.ca/"
driver.get(url) # open the website

driver.maximize_window() # miximize window

driver.find_element("xpath", "//input[@id=\'MainContent_physicianSearchView_txtFirstName\']").click() # click on the first name input box

driver.find_element("xpath","//input[@id='MainContent_physicianSearchView_txtFirstName']").send_keys("a") # enter first name
time.sleep(2) # time delay for input to be registered

driver.find_element("xpath", "//input[@id=\'MainContent_physicianSearchView_btnSearch\']").click() # click on the search button
time.sleep(7) # time delay for results to load 
```

```python
# import time
# import pandas as pd
# from selenium import webdriver
# from bs4 import BeautifulSoup

# # Initialize WebDriver (make sure to replace with your own WebDriver path)
# driver = webdriver.Chrome(executable_path='path_to_chromedriver')

# # Open the website
# driver.get("your_target_url")

# Initialize empty lists to store the data
names = []
locations = []
disciplines = []
genders = []
languages = []

# Define a loop to extract data from multiple pages
for i in range(5):
    # Parse the page source
    soup = BeautifulSoup(driver.page_source, "html.parser")
    
    # Find the results table
    table = soup.find("table", id="MainContent_physicianSearchView_gvResults")
    
    # Find all rows in the table
    trs = table.find_all("tr")
    
    # Extract the data for each row
    for tr in trs[1:]:  # Skip the header row
        soup_tr = BeautifulSoup(str(tr), 'html.parser')
        
        # Extract Name
        name_tag = soup_tr.find('td', class_='status4')
        name = name_tag.get_text(strip=True) if name_tag else None
        names.append(name)
        
        # Extract Location
        location_tag = soup_tr.find_all('td')[1] if len(soup_tr.find_all('td')) > 1 else None
        location = location_tag.get_text(strip=True) if location_tag else None
        locations.append(location)
        
        # Extract Practice Discipline
        discipline_tag = soup_tr.find_all('td')[2] if len(soup_tr.find_all('td')) > 2 else None
        discipline = discipline_tag.get_text(strip=True) if discipline_tag else None
        disciplines.append(discipline)
        
        # Extract Gender
        gender_tag = soup_tr.find_all('td')[3] if len(soup_tr.find_all('td')) > 3 else None
        gender = gender_tag.get_text(strip=True) if gender_tag else None
        genders.append(gender)
        
        # Extract Languages
        language_tag = soup_tr.find_all('td')[4] if len(soup_tr.find_all('td')) > 4 else None
        language = language_tag.get_text(strip=True) if language_tag else None
        languages.append(language)
    
    # Check if the "Next" button is available and click it
    try:
        next_button = driver.find_element("xpath", "//input[@value='Next']")  # Modify as per the actual button element
        if next_button.is_enabled():  # Check if the button is enabled
            next_button.click()  # Click the "Next" button
            time.sleep(4)  # Wait for the next page to load
        else:
            break  # If the button is not available, break the loop
    except Exception as e:
        print(f"Error while trying to click next: {e}")
        break

# Create a DataFrame from the extracted data
data = {
    'Name': names,
    'Location': locations,
    'Practice Discipline': disciplines,
    'Gender': genders,
    'Languages': languages
}

df = pd.DataFrame(data)
df
```
