```python
import requests
from bs4 import BeautifulSoup
from selenium import webdriver
import time
import pandas as pd

driver=webdriver.Chrome()
url='https://getlatka.com/saas-companies?page=1'
driver.get(url)
time.sleep(5)
driver.find_element('xpath','//button[@class="newsletter-popup__close"]').click()
time.sleep(3)
soup=BeautifulSoup(driver.page_source,'html')
#soup.find_all('tr',class_='data-table_row__aX_dq ')
#print(trs)
```

```python
table=soup.find('table',class_='data-table_table__SwBLY')
trs=table.find_all('tr',class_='data-table_row__aX_dq')

```

```python
data_list = []

for tr in trs:
    dic = {
        "company_name": "",
        "company_page": "",
        "website": "",
        "linkedin": "",
        "facebook": "",
        "twitter": "",
        "crunchbase": "",
        "founder_name": "",
        "founder_page": "",
        "youtube": "",
        "country": ""
    }

    # ------------ COMPANY NAME + COMPANY PAGE ------------
    name_tag = tr.find("a", class_="cells_link__PfQot")
    if name_tag:
        dic["company_name"] = name_tag.text.strip()
        dic["company_page"] = "https://getlatka.com" + name_tag.get("href", "")

    # ------------ SOCIAL + WEBSITE + LINKEDIN + FACEBOOK + TWITTER + CRUNCHBASE --------
    social_links = tr.find_all("a")
    for a in social_links:
        href = a.get("href", "").lower()

        # Website (general)
        if ("http" in href or "//" in href) and "linkedin" not in href and "facebook" not in href and "twitter" not in href and "crunchbase" not in href:
            dic["website"] = href

        # LinkedIn
        if "linkedin.com" in href:
            dic["linkedin"] = href

        # Facebook
        if "facebook.com" in href:
            dic["facebook"] = href

        # Twitter / X
        if "twitter.com" in href or "x.com" in href:
            dic["twitter"] = href

        # Crunchbase
        if "crunchbase.com" in href:
            dic["crunchbase"] = href

    # ------------ FOUNDER NAME + FOUNDER PAGE ------------
    founder_tag = tr.find("a", class_="cells_name__pBrsJ")
    if founder_tag:
        dic["founder_name"] = founder_tag.text.strip()
        dic["founder_page"] = "https://getlatka.com" + founder_tag.get("href", "")

    # ------------ YOUTUBE LINK ------------
    yt = tr.find("a", class_="youtube-link")
    if yt:
        dic["youtube"] = yt.get("href", "")

    # ------------ COUNTRY ------------
    country_tag = tr.find("a", href=lambda x: x and "/countries" in x)
    if country_tag:
        dic["country"] = country_tag.text.strip()

    data_list.append(dic)

# ----- CREATE DATAFRAME -----
df = pd.DataFrame(data_list)
df
```

```python
df.to_csv('getlatka_page_1.csv',index=False)
```
