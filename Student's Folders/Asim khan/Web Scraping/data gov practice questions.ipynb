```python
import requests
from bs4 import BeautifulSoup

# Step 1: Website ka link
url = "https://catalog.data.gov/dataset?q=&sort=metadata_created+desc"

# Step 2: Website se data lena
page = requests.get(url)

# Step 3: HTML ko BeautifulSoup se read karna
soup = BeautifulSoup(page.text, "html.parser")

# Step 4: Har dataset ka main div find karna
all_data = soup.find_all("div", class_="dataset-content")

# Step 5: Har dataset ke liye loop
for item in all_data:
    # Heading (h3 ke andar <a> tag hota hai)
    h3_tag = item.find("h3").find("a")
    heading = h3_tag.text.strip()

    # Paragraph (District of Columbia ... wala part)
    p_tag = item.find("p")
    paragraph = p_tag.text.strip()

    # Div (HTML, CSV, GeoJSON ... wala part)
    div_tag = p_tag.find_next_sibling("div")
    div_text = div_tag.text.strip()

    # Print output
    print(heading)
    print(f"{paragraph}{div_text}")
    print("-" * 60)
```
