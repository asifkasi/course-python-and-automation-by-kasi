## 2.1 Extract First movie details and save into csv.(movie_title, movie_index, year, ratings, votes, link) 

```python
# Required libraries import kar rahe hain
import requests  # Website se data (HTML) fetch karne ke liye
from bs4 import BeautifulSoup  # HTML ko parse (analyze) karne ke liye
import pandas as pd  # Data ko structured form me store/handle karne ke liye

# Request header define kar rahe hain taake website ko lage ye ek normal browser request hai
header = {
    'User-agent': 'Baiduspider'  # Ye ek browser-like user-agent name hai
}

# IMDb Top 250 movies page se HTML content fetch kar rahe hain 
r1 = requests.get('https://www.imdb.com/chart/top/', headers=header)

# Server se response code check kar rahe hain (200 ka matlab success)
r1.status_code
```

```python

# HTML content ko BeautifulSoup object me convert kar rahe hain for parsing
soup = BeautifulSoup(r1.text, 'html.parser')

# IMDb ke HTML structure me pehla <li> element find kar rahe hain jisme movie ki detail hoti hai
web1 = soup.find('li', class_='ipc-metadata-list-summary-item')

# Empty list bana rahe hain jisme har movie ka data (dictionary form me) store hoga
data1 = []

# web1 ke andar ke saare nested elements (li ke andar) par loop chala rahe hain
for frt_li in web1:
    
    # Har movie ke liye ek dictionary bana rahe hain jisme sab fields (keys) define kar rahe hain
    dic1 = {
        'tittle': '',
        'index': '',
        'url': '',
        'year': '',
        'hours': '',
        'rating': '',
        'votes': '',
    }

    # Movie ka title extract kar rahe hain h3 tag se, aur "1. The Shawshank Redemption" ko split kar rahe hain
    dic1['tittle'] = frt_li.find('h3').text.strip().split('. ')[-1]

    # Movie ka rank (index number) extract kar rahe hain, jo '.' se pehle hota hai
    dic1['index'] = frt_li.find('h3').text.strip().split('. ')[0]

    # Movie ke link (URL) ko extract kar rahe hain <a> tag ke href attribute se
    dic1['url'] = frt_li.find('a')['href']

    # Year aur duration dono span tags me diye hote hain; pehla span year hota hai
    dic1['year'] = frt_li.find_all('span', class_='cli-title-metadata-item')[0].text.strip()

    # Dusra span duration (hours/minutes) hota hai
    dic1['hours'] = frt_li.find_all('span', class_='cli-title-metadata-item')[1].text.strip()

    # IMDb rating extract kar rahe hain jahan rating star icon hota hai
    dic1['rating'] = frt_li.find('span', class_='ipc-rating-star--rating').text.strip()

    # Total votes extract kar rahe hain aur brackets [ ] hata rahe hain using slicing [1:-1]
    dic1['votes'] = frt_li.find('span', class_='ipc-rating-star--voteCount').text.strip()[1:-1]

    # Ek movie ka dictionary data print kar rahe hain console par
    print(dic1)

    # Us movie ke data ko list me append kar rahe hain
    data1.append(dic1)
```

```python
df=pd.DataFrame(data1)
df
```

```python
df.to_csv("IMDB q1.csv",index=False)
```

## 2.1 Extract All 250 movies details and save into csv. (movie_title, movie_index, year, ratings, votes, link)

```python
# Required libraries import kar rahe hain
import requests               # Website se HTML data fetch karne ke liye
from bs4 import BeautifulSoup # HTML page ko parse (analyze) karne ke liye
import pandas as pd           # Data ko table form me store aur handle karne ke liye

# Ek header dictionary bana rahe hain jisse website ko lage ke ye ek browser se aayi request hai
header = {
    'User-agent': 'Baiduspider'  # User-agent specify kar rahe hain (bot ya browser name)
}

# IMDb ke Top 250 movies page se HTML content fetch kar rahe hain using GET request
r = requests.get('https://www.imdb.com/chart/top/', headers=header)

# Server ka response status code check kar rahe hain (200 ka matlab request successful hai)
r.status_code
```

```python
# Fetched HTML content ko BeautifulSoup object me convert kar rahe hain for parsing
soup = BeautifulSoup(r.text, 'html.parser')

# IMDb page ke sabhi <li> elements dhoond rahe hain jinke class me movie ka data hota hai
web = soup.find_all('li', class_='ipc-metadata-list-summary-item')

# Ek empty list bana rahe hain jisme har movie ka dictionary data store hoga
data = []

# Har ek movie ke li element par loop chala rahe hain
for li in web:
    
    # Ek empty dictionary bana rahe hain har movie ke details store karne ke liye
    dic = {
        'tittle': '',
        'index': '',
        'url': '',
        'year': '',
        'hours': '',
        'rating': '',
        'votes': '',
    }

    # <h3> tag se title text nikal rahe hain aur "1. The Shawshank Redemption" ko split karke title le rahe hain
    dic['tittle'] = li.find('h3').text.strip().split('. ')[-1]

    # Usi text me se movie ka rank (1, 2, 3, ...) nikal rahe hain
    dic['index'] = li.find('h3').text.strip().split('. ')[0]

    # <a> tag ke href attribute se movie ka link le rahe hain
    dic['url'] = li.find('a')['href']

    # <span> tags me se pehla span movie ka release year hota hai
    dic['year'] = li.find_all('span', class_='cli-title-metadata-item')[0].text.strip()

    # Dusra span movie ka duration (hours/minutes) hota hai
    dic['hours'] = li.find_all('span', class_='cli-title-metadata-item')[1].text.strip()

    # IMDb rating extract kar rahe hain jahan rating star icon hota hai
    dic['rating'] = li.find('span', class_='ipc-rating-star--rating').text.strip()

    # Votes count extract kar rahe hain aur parentheses "(" ")" hata rahe hain [1:-1] slicing se
    dic['votes'] = li.find('span', class_='ipc-rating-star--voteCount').text.strip()[1:-1]

    # Dictionary print kar rahe hain taake console par output dikhe (ek movie ka data)
    print(dic)

    # Dictionary ko list 'data' me add kar rahe hain for later use (jaise DataFrame me convert karna)
    data.append(dic)
```

```python
# Dictionary list ko DataFrame me convert kar rahe hain
dg=pd.DataFrame(data)
dg
```

```python
# DataFrame ko CSV file me export kar rahe hain
dg.to_csv('imdb q2.csv',index=False)
```
