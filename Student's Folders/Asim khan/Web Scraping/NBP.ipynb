```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
url='https://www.nbp.com.pk/RATESHEET/index.aspx'
r=requests.get(url)
r.status_code
soup=BeautifulSoup(r.text,'html.parser')
table=soup.find('table',id='ctl00_ContentPlaceHolder1_GridView1')
tr=table.find_all('tr')
tr
list=[]
for i in tr:
    dic={
        'pdf_name':'',
        'pdf_url':'',
    }
    dic['pdf_name']=i.find('span').text.strip()
    dic['pdf_url']=i.find('a')['href']
    base_url='https://www.nbp.com.pk/'
    actral_url=base_url+dic['pdf_url']
    dic['pdf_url']=actral_url 
    list.append(dic)
list
df=pd.DataFrame(list)
df
for i in range(0,3):
    url_pdf=df['pdf_url'][i]
    pdf_r=requests.get(url_pdf)
    pdf_r.status_code
    file_name=df['pdf_name'][i]+'.pdf'
    #"D:\GitHub\asifkasi\course-python-and-automation-by-kasi\Student's Folders\Asim khan\Web Scraping\NBP PDF\\"+
    with open(file_name,'wb') as f:
        f.write(pdf_r.content)
    f.close()

```
