# All div find

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
r=requests.get('https://www.scrapethissite.com/pages/')
r.status_code
```

```python
soup=BeautifulSoup(r.text,'html.parser')
print(soup.prettify())
```

```python
website=soup.find_all('div',class_='page')
list=[]
for div in website:
    products_dtl={
        'products_name':"",
        'products_para':"",
        'product_url':"",
    }
    products_dtl['products_name']=div.find('a').text
    products_dtl['products_para']=div.find('p').text.strip()
    products_dtl['product_url']=div.find('a')['href']
    print(products_dtl)
    list.append(products_dtl)
```

```python
csv=pd.DataFrame(list)
csv
```

```python
csv.to_csv('list of product.csv',index=False)
```

## 3.1 Extract Table of first page & save into csv file. (Team Name, Year, Wins, Losses, OT Losses, Win %, Goals For (GF), Goals Against (GA), + / -)

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
r=requests.get('https://www.scrapethissite.com/pages/forms/')
r.status_code
```

```python
soup=BeautifulSoup(r.text,'html.parser')
web=soup.find_all('tr',class_='team')
list=[]
for table in web:
    dic={
        'Team Name':'',
        'Year':'',
        'Wins':'',
        'Losses':'',
        'OT Losses':'Nill',
        'Win %':'',
        'Goals For (GF)':'',
        'Goals Against (GA)':'',
        '+ / -':'',
    }
    dic['Team Name']=table.find('td',class_='name').text.strip()
    dic['Year']=table.find('td',class_='year').text.strip()
    dic['Wins']=table.find('td',class_='wins').text.strip()
    dic['Losses']=table.find('td',class_='losses').text.strip()
    dic['Win %']=table.find('td',class_='pct').text.strip()
    dic['Goals For (GF)']=table.find('td',class_='gf').text.strip()
    dic['Goals Against (GA)']=table.find('td',class_='ga').text.strip()
    dic['+ / -']=table.find('td',class_='diff').text.strip()
    #print(dic)
    list.append(dic)
print(list)
```

```python
df=pd.DataFrame(list)
df
```

```python
df.to_csv('scrapethissite_q1.csv',index=False)
```
