## 3. Write a Python program to that retrieves an arbitary Wikipedia page of "Python" and creates a list of links on that page.

```python
import requests
from bs4 import BeautifulSoup
headers={
    "User-agent": "Orthogaffe"
}
web3=requests.get('https://en.wikipedia.org/wiki/Python',headers=headers)
web3.status_code
```

```python
soup=BeautifulSoup(web3.text,'html.parser')
links=soup.find_all('a')
for i in links:
    print(i['href'])
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
headers={
    'User-agent':'IsraBot'
}
url = 'https://en.wikipedia.org/wiki/Peter_Jeffrey_(RAAF_officer)'
r = requests.get(url,headers=headers)
soup = BeautifulSoup(r.text,'html.parser')

images = soup.find_all('img')

list_img = []

for i in images:
    dic = {
        'image_url':''
    }

    src = i['src'].replace('//', 'https://').replace('/wiki', 'https://en.wikipedia.org/wiki')

    dic['image_url'] = src
    list_img.append(dic)

df = pd.DataFrame(list_img)
df
```
