```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

r = requests.get('https://books.toscrape.com/')
soup = BeautifulSoup(r.text, 'html.parser')

books = soup.find_all('li', class_='col-xs-6 col-sm-4 col-md-3 col-lg-3')

data = []

for book in books:
    # Rating extract karna
    rating_tag = book.find('p', class_='star-rating')
    rating = rating_tag['class'][1]
    # Price extract karna
    price_tag = book.find('p', class_='price_color')
    price = price_tag.text.strip()

    data.append({'Rating': rating, 'Price': price})

# DataFrame mein store karna (optional)
df = pd.DataFrame(data)
print(df)

```

```python
df.to_csv('book to scrap.csv',index=False)
```

## 8.1 Extract first product details save into csv file. e.g# A Light in the..(Product_name(full name click via clicking on product url),Product_Description, UPC, Product_Type, Price_excl_tax, Price_incl_tax, Tax, Availability_numbers, Number_of_reviews, product_img, product_reviews_star_no, product_price, product_availabililty)

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
main_url='https://books.toscrape.com/'
r=requests.get(main_url)
r.status_code
```

```python
soup=BeautifulSoup(r.text,'html.parser')
products=soup.find_all('li',class_='col-xs-6 col-sm-4 col-md-3 col-lg-3')
list = []
for fst_pro in products:
    dict_pro = {
    "Product_name":"",
    "Product_url":"",
    "product_img":"",
    "product_price":"",
    "product_availabililty":""
    }
    dict_pro["Product_name"]= fst_pro.find("h3").text
    dict_pro["product_price"]= fst_pro.find("p",class_ = "price_color").text
    dict_pro["product_availabililty"]= fst_pro.find("p",class_ = "instock availability").text.strip()
    dict_pro["product_img"]= fst_pro.find("img")["src"]
    dict_pro["Product_url"]= fst_pro.find("a")["href"]
    list.append(dict_pro)

df  = pd.DataFrame(list)
df
```
