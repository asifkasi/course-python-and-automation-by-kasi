```python
# 1.  Print a square pattern of `*` of size `n x n`. 
#    -  Input:  `n = 3`
#    -  Expected Output: 
      
#       ***
#       ***
#       ***
n=int(input("enter any number"))
for i in range(1,n+1):
    for j in range(1,n+1):
        print("*",end="")
    print()
```

```python
# 2.  Print a right-angled triangle pattern of `*`. 
#    -  Input:  `n = 3`
#    -  Expected Output: 
      
#       *
#       **
#       ***
n=int(input("enter any number"))
for i in range(1,n+1):
    for j in range(i):
        print("*",end="")
    print()
```

```python
# 3.  Print an inverted right-angled triangle pattern of `*`. 
#    -  Input:  `n = 3`
#    -  Expected Output: 
      
#       ***
#       **
#       *
n=int(input("enter any number"))
for i in range(n,0,-1):
    for j in range(i):
        print("*",end="")
    print()
```

```python
# 8.  Print a number pattern where each row contains the same number (e.g., 1, 22, 333, etc.). 
#    -  Input:  `n = 3`
#    -  Expected Output: 
      
#      1
#      22
#      333
n=int(input("enter a number"))
for i in range(1,n+1):
    for j in range(i):
        print(i,end="")
    print()
```

```python
# 9.  Print a pattern of increasing numbers (e.g., 1, 12, 123, etc.). 
#    -  Input:  `n = 3`
#    -  Expected Output: 
      
#      1
#      12
#      123
n=int(input("enter a number"))
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

```python
# 10.  Print a pattern of decreasing numbers (e.g., 321, 32, 3, etc.). 
#     -  Input:  `n = 3`
#     -  Expected Output: 
       
#       321
#       32
#       3
n=int(input("enter a number"))
for i in range(1,n+1):
    for j in range(n,i-1,-1):
        print(j,end="")
    print()
```

```python
# 11.  Print a multiplication table up to `n x n`. 
#     -  Input:  `n = 3`
#     -  Expected Output: 
       
#       1  2  3
#       2  4  6
#       3  6  9
n=int(input("enter a number"))
for i in range(1,n+1):
    for j in range(1,n-1):
        print(f"{i} {i*2} {i*3}",end=" ")
    print()
```

```python
# 12.  Print a pattern of odd numbers in rows (e.g., 1, 3, 5, etc.). 
#     -  Input:  `n = 3`
#     -  Expected Output: 
       
#       1
#       3 5
#       7 9 11
n=int(input("enter a number"))
sum=1
for i in range(1,n+1):
    for j in range(0,i):
        print(sum,end=" ")
        sum=sum+2
       
    print()
```

```python
# 13.  Print a pattern of even numbers in rows (e.g., 2, 4, 6, etc.). 
#     -  Input:  `n = 3`
#     -  Expected Output: 
       
#       2
#       4 6
#       8 10 12
n=int(input("enter a number"))
sum=2
for i in range(1,n+1):
    for j in range(i):
        print(sum,end=" ")
        sum=sum+2
    print()
```

```python
# 14.  Print a pattern of alternating 1s and 0s in a square grid. 
#     -  Input:  `n = 3`
#     -  Expected Output: 
       
#       1 0 1
#       0 1 0
#       1 0 1
n=int(input("enter a number"))
for i in range(1,n+1):
    for j in range(i,i+3):
        if j%2==1:
            print("1",end=" ")
        else:
            print("0",end=" ")
    print()
```
