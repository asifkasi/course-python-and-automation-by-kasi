```python
# student_management.py

students = {}

def add_student(roll, name, marks):
    students[roll] = {"name": name, "marks": marks}

def search_student(roll):
    return students.get(roll, "Student not found!")

def display_students():
    for roll, data in students.items():
        print(f"Roll: {roll}, Name: {data['name']}, Marks: {data['marks']}")

while True:
    print("\n1. Add Student\n2. Search Student\n3. Display All\n4. Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        r = input("Roll No: ")
        n = input("Name: ")
        m = input("Marks: ")
        add_student(r, n, m)

    elif choice == "2":
        r = input("Enter Roll No: ")
        print(search_student(r))

    elif choice == "3":
        display_students()

    elif choice == "4":
        break
```
