```python
#take 5 int input and find the count of even and odd
even=0
odd=0
evenlist=[]
oddlist=[]
for i in range(5):
    num=int(input("enter number"))
    if num%2==0 and num!=0:
        even=even+1
        evenlist.append(num)
    elif num%2==1:
        odd=odd+1
        oddlist.append(num)
print("input your even number is",evenlist)
print("input your odd number is",oddlist)
print("count of ever number is",even)
print("count of odd number is",odd)
```
