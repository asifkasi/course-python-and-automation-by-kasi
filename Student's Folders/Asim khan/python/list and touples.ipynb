```python
# this is all methods in python
list=[38,48,59,23,48,29,88,99]
#append method (append method is add any number in list for example [1,2,3] append(4) then [1,2,3,4])
list.append(66)
print(list)
#sort method (sort method like a ascending order for example [2,1,3] then [1,2,3])
list.sort()
print(list)
#sort reversed (sort reversed like a decending order for ex [1,2,3] then [3,2,1])
list.sort(reverse=True)
print(list)
# reverse method (reverse method all value is reverse for ex [2,1,3]then[3,1,2])
list.reverse()
print(list)
#insert method(add any number in darmyan me)
list.insert(3,22)
print(list)
```

```python
#remove method (remove any number you can see below)
list=[38,48,59,23,48,29,88,99]
list.remove(23)
print(list)
#pop method (pop method is same as remove method but the difference between we use pop method written index)
list.pop(4)
print(list)
```

```python
students = []
def add_student():
    name = input("Enter student name: ")
    grade = float(input("Enter student grade: "))
    student = (name, grade)  # Use a tuple
    students.append(student)
    print(f"{name} added successfully!\n")
def display_students():
    if not students:
        print("No students added yet.\n")
        return
    print("\nAll Students:")
    for name, grade in students:
        print(f"{name} - Grade: {grade}")
    print()
def top_student():
    if not students:
        print("No students to evaluate.\n")
        return
    top = max(students, key=lambda x: x[1])  # x[1] is the grade
    print(f"Top Student: {top[0]} with Grade: {top[1]}\n")
def search_student():
    search_name = input("Enter student name to search: ")
    found = False
    for name, grade in students:
        if name.lower() == search_name.lower():
            print(f"{name} found with Grade: {grade}\n")
            found = True
            break
    if not found:
        print("Student not found.\n")
def menu():
    while True:
        print("=== Student Grade Manager ===")
        print("1. Add Student")
        print("2. Display Students")
        print("3. Show Top Student")
        print("4. Search Student")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")
        print()

        if choice == '1':
            add_student()
        elif choice == '2':
            display_students()
        elif choice == '3':
            top_student()
        elif choice == '4':
            search_student()
        elif choice == '5':
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Try again.\n")

menu()
```
