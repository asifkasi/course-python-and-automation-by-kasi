```python
# Print a pattern of increasing letters (e.g., A, AB, ABC, etc.). 
#     -  Input:  `n = 3`
#     -  Expected Output: 
       
#       A
#       AB
#       ABC

n=3 
for i in range(1,n+1):
    for j in range(65,65+i):
        print(chr(j), end="")
    print()
```

```python
# Print a pattern where each row contains letters and numbers combined. 
#     -  Input:  `n = 3`
#     -  Expected Output: 
       
#       A1
#       B2 C3
#       D4 E5 F6
n=int(input("enter a number"))
num=1
for i in range(1,n+1):
    for j in range(i):
        print(f"{chr(num+64)}{num}", end=" ")
        num+=1
    print()
```

```python
# Print a diagonal pattern within a square grid. 
#     -  Input:  `n = 3`
#     -  Expected Output: 
       
#       *  
#        * 
#         *
n=3
for i in range(n+1):
    for j in range(i):
        print(" ", end=" ")
    print("*")
```
