```python
# library.py

library = {}

def add_book(id, title, author):
    library[id] = {"title": title, "author": author, "available": True}

def borrow_book(id):
    if id in library and library[id]["available"]:
        library[id]["available"] = False
        print(f"You borrowed '{library[id]['title']}'")
    else:
        print("Book not available!")

def return_book(id):
    if id in library and not library[id]["available"]:
        library[id]["available"] = True
        print(f"Book '{library[id]['title']}' returned!")
    else:
        print("Invalid book ID!")

def show_books():
    for id, info in library.items():
        status = "Available" if info["available"] else "Borrowed"
        print(f"{id}: {info['title']} by {info['author']} - {status}")

while True:
    print("\n1. Add Book\n2. Borrow Book\n3. Return Book\n4. Show Books\n5. Exit")
    ch = input("Enter choice: ")

    if ch == "1":
        i = input("ID: ")
        t = input("Title: ")
        a = input("Author: ")
        add_book(i, t, a)

    elif ch == "2":
        i = input("Enter Book ID: ")
        borrow_book(i)

    elif ch == "3":
        i = input("Enter Book ID: ")
        return_book(i)

    elif ch == "4":
        show_books()

    elif ch == "5":
        break
```
