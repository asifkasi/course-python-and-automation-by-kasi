# 5. https://www.scrapethissite.com/pages/simple/ (coutry_name, Capital, Population, Area)

```python
# from urllib.request import urlopen
import requests
from bs4 import BeautifulSoup
import re


header={
    "User-Agent": "Baiduspider"
}

html = requests.get('https://www.scrapethissite.com/pages/simple/',headers=header)
# html = requests.get('https://www.imdb.com/chart/top/')

html.status_code
```

```python
soup = BeautifulSoup(html.text, "html")
print(soup.prettify())
```

# Andorra

```python
divs = soup.find_all("div",class_="col-md-4 country")
print(divs)
len(divs)
```

```python
country_info = {
    "country_name":"",
    "country_capital":"",
    "country_population":"",
    "country_area":""

}
country_info
```

```python
list_1 = []

for div in divs[0:249]:
    country_info = {
    "country_name":"",
    "country_capital":"",
    "country_population":"",
    "country_area":""
    }
    

    country_info["country_name"] = div.find("h3").text.strip()
    country_info["country_capital"] = div.find("span",class_="country-capital").text.strip()
    country_info["country_population"] = div.find("span",class_="country-population").text.strip()
    country_info["country_area"] = div.find("span",class_="country-area").text.strip()



    list_1.append(country_info)
print(list_1)

```

```python
import pandas as pd
list = []
list.append(country_info)
df = pd.DataFrame(list_1)
df
```

```python
df.to_csv("countryoo_info.csv",index=False)
```
