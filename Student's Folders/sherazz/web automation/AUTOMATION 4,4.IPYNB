```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import time
driver = webdriver.Chrome()
driver.get("https://webscraper.io/test-sites/e-commerce/ajax/computers/laptops")
time.sleep(3)
driver.find_element("xpath" , "//button [@ class='acceptCookies']").click()


for btn in driver.find_elements("xpath" , "//button [@ class = 'btn btn-default page-link page']"):
    if btn.text == '2':
        btn.click()
        break

time.sleep(6)

soup = BeautifulSoup(driver.page_source , "html")
print(soup.prettify())
```

```python
row = soup.find_all("div" , class_ = "col-md-4 col-xl-4 col-lg-4")
row
```

```python
import requests
dl = []

for i in range(6):
    p_url = row[i].find("a")["href"]
    b_url = "https://webscraper.io"
    act_url = b_url + p_url

    r1 = requests.get(act_url)
    soup1 = BeautifulSoup (r1.text , "html.parser")

    data = {} 
    data["Product_Name"] = soup1.find("h4" ,class_ = "title card-title").text
    data["Product_Link"] = act_url
    data["Product_Image"] = soup1.find("img" ,class_ = "image img-fluid img-responsive")["src"]
    data["Product_Price"] = soup1.find("h4" , class_ = "price float-end pull-right").text.strip()
    data["Product_Details"] = soup1.find("p" , class_ = "description card-text").text 
    data["Product_Review_No"] = soup1.find("p" , class_ = "review-count").text.strip()
    
    dl.append(data)
dl
```

```python
df=pd.DataFrame(dl)
df
```

```python
df.to_csv("question_no_4.csv", index=False)    
df  
```
