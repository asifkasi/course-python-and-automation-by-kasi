```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import time
import pandas as pd

driver = webdriver.Chrome()
url = "https://www.scrapethissite.com/pages/frames/"
driver.get(url)




driver.switch_to.frame(0)


driver.find_element(By.XPATH, "//h3").text

soup = BeautifulSoup(driver.page_source , "html")
soup
divs = soup.find_all("div" , class_ = "col-md-4 turtle-family-card")
len(divs)
print(divs[0].prettify())
div = divs[0]


dict = {
    "Name" : "",
    "url" : "",
    "Image_url" : "",

}

base_url = "https://www.scrapethissite.com"
dict["Name"] = div.find("h3").text.strip()
dict["url"] = base_url + div.find("a")["href"]
dict["Image_url"] = base_url + div.find("img")["src"]

dict
df = pd.DataFrame([dict])
df
df.to_csv("question_no_2(frames).csv", index=False)    
df  
```
