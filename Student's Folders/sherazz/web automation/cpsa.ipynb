```python
import requests
from bs4 import BeautifulSoup
import time
import pandas
from selenium import webdriver
driver = webdriver.Chrome()
driver.get("https://search.cpsa.ca/")


driver.maximize_window()

```

```python
driver.find_element("xpath", "//input[@id='MainContent_physicianSearchView_txtFirstName']").click()
driver.find_element("xpath", "//input[@id='MainContent_physicianSearchView_txtFirstName']").send_keys("a")
time.sleep(2)
driver.find_element("xpath", "//input[@id='MainContent_physicianSearchView_btnSearch']").click()
time.sleep(7)
```
