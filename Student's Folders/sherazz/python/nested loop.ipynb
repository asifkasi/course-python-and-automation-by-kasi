```python
# ============nested loop question no 1=============
# n= "*"
# for i in range(1,4):
#     for j in range(1):
#         print(n,n,n)
# # print()    


# Input from user
n = int(input("Enter the size of square: "))

# Loop for rows
for i in range(n):
    # Loop for columns
    for j in range(n):
        print("*", end="")  # print * without new line
    print()  # move to next line
```

```python
# ============nested loop question no 2=============


# Input from user
n = int(input("Enter the size of triangle: "))

# Loop for rows
for i in range(1, n + 1 ):
    # Loop for stars
    for j in range(i):
        print("*", end="")
    print()  # move to next line





```
