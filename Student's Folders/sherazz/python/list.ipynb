```python
# array ka concept python mei list ke zaryee karty hai
# There are 4 type of data collection in python :
# 1) list 2)tuple 3)set 4)dictinonry
# list mei hum value ko change kar sakty hai for example : insert ,delete ,etc
# list mei hum acending se start kry gy to index 0 dengy aur agr decending se start kry gy to index -1 dengy


```

```python
list = ["sheraz","khani","ali","luqman","fahad"]
# [0] isko kahty hai index aur index start 0 se hota hai agar mei 1 donga to 2nd name print kar dega.
print(list [0]) 
# [-1] ye dosri side se start kry ga aur decending mei index -1 se start hota hai
# is pore concept ko slicing kahty hai
print(list[-1])
# [0:3] isko kahty hai rang dena ke 0 se 3 tak print karo
print(list [0:3])
# [:2] iska matlab hai ke ye start 0 se kry ga aur 2 tak jaeega 
print(list [:2])
# [2:] iska matlab hota hai ke 2 se start kry ga aur last tak jaeega 
print(list[2:])
print(list[-3:])
# [::-3 ] iska matlab hota hai ke negative se start kry gab aur 3 ke baad print kry ga
print(list[::-3])
```
