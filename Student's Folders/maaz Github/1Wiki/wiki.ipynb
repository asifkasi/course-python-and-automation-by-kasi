## 1. Write a Python program to extract and display all the header tags from en.wikipedia.org/wiki/Main_Page

```python
import requests
from bs4 import BeautifulSoup
url = "https://en.wikipedia.org/wiki/Main_Page"
r = requests.get(url, headers={ 
    "User-Agent": "IsraBot"})
soup = BeautifulSoup(r.text, 'html.parser')
print(soup.prettify())
headers = soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6', 'h7'])
for header in headers:
    print(header.text)
```
