## 3.1 Extract first product details save into csv file

```python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://books.toscrape.com/")
print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')

print(soup.prettify())
```

```python
pro = soup.find("li", class_ ="col-xs-6 col-sm-4 col-md-3 col-lg-3")
pro
```

## (Product_name,product_img, product_reviews_star_no, product_price, product_availabililty)

```python
product = {

    "product_name":"",
    "product_img":"",
    "product_price":"",
    "product_availabililty":"", 
   
}

product["product_name"]= pro.find('h3').text.strip()
product["product_img"]= pro.find('img')["src"]
product["product_price"]= pro.find('p', class_= "price_color").text.strip()
product["product_availabililty"]= pro.find('p', class_= "instock availability").text.strip()
product
```

```python
import pandas as pd
list = []
list.append(product)
df = pd.DataFrame(list)
df
```

```python
df.to_csv("book_toscrape q_3.1.csv", index=False)
```
