## 6.1 Extract first product details save into csv file. e.g# Asus VivoBook...

```python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops")
print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, 'html.parser')

print(soup.prettify())
```

```python
product_div = soup.find("div", class_ = "product-wrapper card-body")
product_div
```

```python
product = {

    "product_name":"",
    "product_link":"",
    "product_img":"",
    "product_price":"",
    "product_details":"", 
    "product_reviews_number":"",
    "product_reviews_star_number":""
}
product
product["product_name"]= product_div.find('a', class_= "title").text.strip()
product["product_link"]= product_div.find('a', class_= "title")["href"]
product["product_img"]= product_div.find('img', class_= "img-fluid card-img-top image img-responsive")["src"]
product["product_price"]= product_div.find('h4', class_= "price float-end card-title pull-right").text.strip()
product["product_details"]= product_div.find('p', class_= "description card-text").text.strip()
product["product_reviews_number"]= product_div.find('p', class_= "review-count float-end").text.strip()
product
```

```python
import pandas as pd
list = []
list.append(product)
df = pd.DataFrame(list)
df
```
