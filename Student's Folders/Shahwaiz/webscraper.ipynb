# 6.1 Extract first product details save into csv file. e.g# Asus VivoBook X4

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
url="https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops"
r=requests.get(url)
r.status_code
```

```python
soup=BeautifulSoup(r.text)
soup
```

```python
first1=soup.find("div",class_="col-md-4 col-xl-4 col-lg-4")
first1
```

```python
product_name=first1.find("a",class_="title").text.strip()
product_name
```

```python
product_img=first1.find("img",class_="img-responsive").get("src")
product_img
```

```python
product_details=first1.find("div",class_="caption").find("p",class_="description").text.strip()
product_details
```

```python
product_price=first1.find("h4",class_="price float-end card-title pull-right").text.strip()
product_price
```

```python
product_rating=first1.find("div",class_="ratings").text.strip()
product_rating
```

```python
dl = []
dict={
    "Product Name":"",
    "Product Image":"",
    "Product Details":"",
    "Product Price":"",
    "Product Rating":""
}
dict["Product Name"]=first1.find("a",class_="title").text.strip()
dict["Product Image"]= first1.find("img",class_="img-responsive").get("src")
dict["Product Details"]=first1.find("div",class_="caption").find("p",class_="description").text.strip()
dict["Product Price"]=first1.find("h4",class_="price float-end card-title pull-right").text.strip()
dict["Product Rating"]=first1.find("div",class_="ratings").text.strip()
dl.append(dict)
dl
```

```python
import pandas as pd
df=pd.DataFrame(dl)
df
```

```python
df.to_csv("web_scraper_q_6_1.csv",index=False)
```
