```python
import requests
from bs4 import BeautifulSoup
import pandas as pd 

url = "https://www.scrapethissite.com/pages/"
r=requests.get(url)
r.status_code
print(r.text)
```

```python
soup=BeautifulSoup(r.text)
soup
```

```python
divs=soup.find("div",class_=("page"))
divs
```

```python
print("divs: ",divs.find("p").text.strip())
```

```python
print("divs: ",divs.find("a")["href"])
```

```python
dist={
    "header":"",
    "para":"",
    "link":"",

}
list=[]
dist["header"]=divs.find("h3").text.strip()
dist["para"]=divs.find("p").text.strip()  
dist["link"]=divs.find("a")["href"]
list.append(dist.copy())
list
```

```python
df=pd.DataFrame(list)
df
```

```python
df.to_csv("data of scrapthissite.csv",index=False)
```

```python

import requests
from bs4 import BeautifulSoup
import pandas as pd 

url = "https://www.scrapethissite.com/pages/"
r=requests.get(url)
r.status_code
print(r.text)
```

```python
divs=soup.find_all("div",class_=("page"))
len(divs)
```

```python
list = []
for i in divs:

    # print(i.text)
    dist["header"]=i.find("h3").text.strip()
    dist["para"]=i.find("p").text.strip()  
    dist["link"]=i.find("a")["href"]
    list.append(dist.copy())
list
```

```python
df=pd.DataFrame(list)
df
```

```python
df.to_csv("data of scrapthissite2.csv",index=False)
```
