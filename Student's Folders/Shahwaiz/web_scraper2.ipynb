#	7.1 Extract first product details save into csv file. e.g# Packard 255 G2

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
url="https://webscraper.io/test-sites/e-commerce/static/computers/laptops"
r=requests.get(url)
r.status_code


```

```python
soup=BeautifulSoup(r.text,"html")
soup
```

```python
First_product=soup.find("div",class_="col-md-4 col-xl-4 col-lg-4")
First_product
```

```python
First_product.text
```

```python
product_name=First_product.find("a",class_="title").text.strip()
product_name
```

```python
product_link=First_product.find("a",class_="title")["href"]
product_link
```

```python
product_reviews_number=First_product.find("p",class_="review-count float-end").text.strip()
product_reviews_number
```

```python
product_price=First_product.find("h4",class_="price float-end card-title pull-right").text.strip()
product_price
```

```python
product_details=First_product.find('p').text

product_details
```

```python
product_img=First_product.find('img')['src']
product_img
```

```python
list=[]
dict={
    "Product Name":'',
    "Product Link":"",
    "Product Reviews Number":"",
    "Product Price":"",
    "product_details":"",
    "product_img":"",
}
dict["Product Name"]=First_product.find("a",class_="title").text.strip()
dict["Product Link"]=First_product.find("a",class_="title")["href"]
dict["Product Reviews Number"]=First_product.find("p",class_="review-count float-end").text.strip()
dict["Product Price"]=First_product.find("h4",class_="price float-end card-title pull-right").text.strip()
dict["product_details"]=First_product.find('p').text.strip()
dict["product_img"]=First_product.find('img')['src']
list.append(dict)   
list
```

```python
df=pd.DataFrame(list)
df
```

```python
df.to_csv("web_scraper2_q_7_1.csv",index=False)
```

# 7.2 Extract product details from first row & save into csv file. e.g# Packard 255 G2, Aspire E1-510, ThinkPad T540p

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
url="https://webscraper.io/test-sites/e-commerce/static/computers/laptops"
r=requests.get(url)
r.status_code
soup=BeautifulSoup(r.text,"html")
soup
```

```python
first_row=soup.find_all("div",class_="row")
len(first_row)
```

```python
all_products = first_row[1].find_all("div",class_="col-md-4 col-xl-4 col-lg-4")[:3]
len(all_products)
```

```python
list=[]
for i in all_products:
    dict={
        "Product Name":'',
        "Product Link":"",
        "Product Reviews Number":"",
        "Product Price":"",
        "product_details":"",
        "product_img":"",
    }
    dict["Product Name"]=i.find("a",class_="title").text.strip()
    dict["Product Link"]=i.find("a",class_="title")["href"]
    dict["Product Reviews Number"]=i.find("p",class_="review-count float-end").text.strip()
    dict["Product Price"]=i.find("h4",class_="price float-end card-title pull-right").text.strip()
    dict["product_details"]=i.find('p').text.strip()
    dict["product_img"]=i.find('img')['src']
    list.append(dict)   
list    
```

```python
#import pandas as pd
df=pd.DataFrame(list)
df
```

```python
df.to_csv("web_scraper2_q_7_2.csv",index=False)
```
