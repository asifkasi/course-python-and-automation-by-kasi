#	3.1 Extract first product details save into csv file. e.g A Light in the ..

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd 
url="https://books.toscrape.com/"
r=requests.get(url)
r.status_code
```

```python
soup=BeautifulSoup(r.text,'html')
soup
```

```python
first_book=soup.find("li",class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")
first_book
```

```python
book_name=first_book.find("h3").text.strip()
book_name
```

```python
book_img=first_book.find("img",class_="thumbnail")["src"]
book_img
```

```python
#book_review_star_no=first_book.find("p",class_="star-rating Three")
#book_review_star_no
```

```python
book_price=first_book.find("p",class_="price_color").text.strip()
book_price
```

```python
book_availability=first_book.find("p",class_="instock availability").text.strip()
book_availability
```

```python
list=[]
dict={
    "Book Name":book_name,
    "Book Image":book_img,
    "Book Price":book_price,
    "Book Availability":book_availability,
}
dict["Book Name"]=first_book.find("h3").text.strip()
dict["Book Image"]=first_book.find("img",class_="thumbnail")["src"]
dict["Book Price"]= first_book.find("p",class_="price_color").text.strip()
dict["Book Availability"]=first_book.find("p",class_="instock availability").text.strip()
list.append(dict)   
list

```

```python
df=pd.DataFrame(list)
df
```

```python
df.to_csv("book_to_strap_q_3_1.csv",index=False)
```
