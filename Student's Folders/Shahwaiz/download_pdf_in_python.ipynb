```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url="https://www.nbp.com.pk/RATESHEET/index.aspx"
r=requests.get(url)
r.status_code
```

```python
soup=BeautifulSoup(r.text,"html")
tb=soup.find("table",id="ctl00_ContentPlaceHolder1_GridView1")
tb
```

```python

div_pdfs=tb.find_all("tr")
div_pdfs
len(div_pdfs)



```

```python
list=[]
for i in div_pdfs:
    dict={
        "pdf_name":"",
        "pdf_url":"",

        
    }
    dict["pdf_name"]=i.span.text.strip()
    dict["pdf_url"]=i.find("a")["href"]
    list.append(dict)
list

    
```

```python
df=pd.DataFrame(list)
df
```

```python
df.to_csv("pdfs_q_a_5.1.cvs",index=False)
```
