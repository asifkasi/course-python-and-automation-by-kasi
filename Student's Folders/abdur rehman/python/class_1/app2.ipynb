```python
# 17.  Print the first `n` rows of Floyd's triangle. 
#     -  Input:  `n = 5`
num = 1 #to start number for floyd triangle 
n = 5 # to input, for rows 
for i in range(1,n): # outer loop start from 1to n+1 for rows
    for j in range(i): # inner loop start from 0 to i for column
        print(num, end=" ") # print num and `end` kyewoed use for stop line break
        num +=1 # increment into valiable `num`,for Floyd's triangle. 
    print() # break line
```

```python
# 9.  Print a pattern of increasing numbers (e.g., 1, 12, 123, etc.). 
#    -  Input:  `n = 3`
n =3
for i in range(0,n+1):
    for j in range(1,i+1):
        print(j, end="")
    print()
```
