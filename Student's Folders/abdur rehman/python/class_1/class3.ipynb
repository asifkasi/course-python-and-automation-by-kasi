```python
#how to print the reverse of elemant of list.
std = ['umair','shariq','asim','umair','wasim','umair']
std.reverse()
print(std)
```

```python
#how to remove the duplication of an element.
lst = ['umair','shariq','asim','umair','wasim','umair']
unique_lst = list(set(lst))
print(unique_lst)
unique_lst = []
for item in lst:
    if item not in unique_lst:
        unique_lst.append(item)
print(unique_lst)        
```

```python
#how to merge two lists.
list1 = ["a","b"]
list2 = ["c","d"]
merged = list1 + list2
print(merged)
```

```python
#what is set in python.
#A set is an unordered,mutable collection of unique element.
s = {1,2,3,4,5,3}
print(s)
```

```python
#what is tuple in python.
# A tuple is an ordered,immutable collection of elements.
t = (1,2,3)
print(t[1])
```

```python
#what is list f list.
#A list of list is a nested structure where each element is itslef a list.
matrix = [1,2,3],[3,4],[5,6]
print(matrix[0])
print(matrix[0][2])
```

```python
# how to use set,tuple and list of list.
#set use
s = {1,2,3,4,5,3}
print(s)
#tuple use 
t = (1,2,3)
print(t[1])
#list of list use
matrix = [1,2,3],[3,4],[5,6]
print(matrix[0])
print(matrix[0][2])
```

```python
# how to find same element between two individual list.
list1 = [ 1,2,3.4]
list2 = [2,3,4]
common = list(set(list1) & set(list2))
print(common)
```
