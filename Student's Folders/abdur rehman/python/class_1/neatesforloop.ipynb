```python
# 1. Square pattern of *
def square_pattern(n):
    for i in range(n):
        print("*" * n)
```

```python
# 8. Number pattern (row number repeated)
def repeated_number_pattern(n):
    for i in range(1, n + 1):
        print(str(i) * i)
```

```python
# 9. Increasing numbers (1, 12, 123, ...)
def increasing_number_pattern(n):
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            print(j, end="")
        print()
```

```python
# 31. Grid of coordinates
def coordinate_grid(n):
    for i in range(n):
        for j in range(n):
            print(f"({i},{j})", end=" ")
        print()
```

```python
# 10. Decreasing numbers (321, 32, 3, ...)
def decreasing_number_pattern(n):
    for i in range(n):
        for j in range(n, i, -1):
            print(j, end="")
        print()
```
