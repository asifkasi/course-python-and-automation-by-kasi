```python
#how many course enrolled first student.
stddic  = {"001": {"name": "Abdur rehman",
               "f`name" : "shakir",
               "age": 18,
               "course enroll":[" math"]
         },
         "002": {"name": "waqas",
               "f`name" : "yaya",
               "age": 18,
               "course enroll":" phy"
         },
            "003":{"name": "shoaib",
               "f`name" : "nana",
               "age": 18,
               "course enroll":"chy"
            },
               } 

count = len(stddic["001"]["course enroll"])

print("Number of courses enrolled by first student:", count)

```

```python
dic1 = {'a': 200,
        'b': 200,
        'c': 200,
        }
dic2 = {'b': 200,
        'd': 200,
        'a': 200,
        }
dic3 = dic2
for key in dic1.keys():
    if key in dic2.keys():
        dic3[key]= dic1[key] + dic2[key]
    else:
        dic3[key]= dic1[key]    
print(dic3)        
```
