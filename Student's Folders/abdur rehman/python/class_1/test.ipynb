```python
#0.8 Program to compute the sum of consecutive odd numbers between two integers
a = int(input("Input first number of the pair: "))
b = int(input("Input second number of the pair: "))
start = min(a, b)
end = max(a, b)
odd_numbers = []
for i in range(start, end + 1):
    if i % 2 == 1:  
        odd_numbers.append(i)
print("List of odd numbers:")
for num in odd_numbers:
    print(num)
print("Sum =", sum(odd_numbers))
```

```python
#2 Program to find the sum of digits of a number
num = int(input("Input a number: "))
sum_digits = 0
temp = num
while temp > 0:
    digit = temp % 10      
    sum_digits += digit     
    temp = temp // 10       
print(f"The sum of digits of {num} is: {sum_digits}")
```
