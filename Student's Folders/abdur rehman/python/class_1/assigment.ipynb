```python
#16. Write a Python program to combine values in a list of dictionaries.
data = [
    {'item': 'item1', 'amount': 400},
    {'item': 'item2', 'amount': 300},
    {'item': 'item1', 'amount': 750}
]
result = {}
for d in data:
    item = d['item']
    amount = d['amount']
    
    if item in result:
        result[item] += amount
    else:
        result[item] = amount

print(result)
```

```python
# 17. Write a Python program to create a dictionary from a string.
text = "w3resource"
result = {}
for ch in text:
    if ch in result:
        result[ch] += 1   
    else:
        result[ch] = 1    

print(result)
```

```python
#0.8 Program to compute the sum of consecutive odd numbers between two integers
a = int(input("Input first number of the pair: "))
b = int(input("Input second number of the pair: "))
start = min(a, b)
end = max(a, b)
odd_numbers = []
for i in range(start, end + 1):
    if i % 2 == 1:  
        odd_numbers.append(i)
print("List of odd numbers:")
for num in odd_numbers:
    print(num)
print("Sum =", sum(odd_numbers))
```

```python
#2 Program to find the sum of digits of a number
num = int(input("Input a number: "))
sum_digits = 0
temp = num
while temp > 0:
    digit = temp % 10      
    sum_digits += digit     
    temp = temp // 10       
print(f"The sum of digits of {num} is: {sum_digits}")
```

```python
#17 Program to display the number in reverse order
num = int(input("Input a number: "))
reverse_num = 0
temp = num
while temp > 0:
    digit = temp % 10          
    reverse_num = reverse_num * 10 + digit
    temp = temp // 10          
print("The number in reverse order is :", reverse_num)
```
