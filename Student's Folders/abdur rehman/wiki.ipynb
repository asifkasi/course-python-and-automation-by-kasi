```python
# Method 1: Saving data into list
import requests
from bs4 import BeautifulSoup
header={
    "User-Agent": "IsraBot"
}

r=requests.get("https://en.wikipedia.org/wiki/Python",headers=header)
soup=BeautifulSoup(r.text,'html.parser')

sub_headings=soup.find_all('div',class_='mw-heading mw-heading2')


var_list_of_dict = []

for i in sub_headings:
    var=i.find('h2')
    if var:
        # print(var.c+":")
        var_dict_data = {
            "sub_header": "",
            "data_point": "",
            "url": ""
        }
        var_dict_data["sub_header"] = var.text

        ul=i.find_next_sibling('ul')
        if ul:
            for li in ul.find_all('li'):
                # text 
                text = li.get_text(strip=True)
                var_dict_data["data_point"] = text
                #url links 
                link_tag = li.find('a')["href"]
                var_dict_data["url"] = link_tag
                # print(f"  - {text} ({link_tag})")

                var_list_of_dict.append(var_dict_data.copy())
                
                print(var_list_of_dict)
```

```python
# Method 2: Saving data into list

import requests
from bs4 import BeautifulSoup
header={
    "User-Agent": "IsraBot"
}

r=requests.get("https://en.wikipedia.org/wiki/Python",headers=header)
soup=BeautifulSoup(r.text,'html.parser')

sub_headings=soup.find_all('div',class_='mw-heading mw-heading2')


var_list_of_dict = []

for i in sub_headings:
    var=i.find('h2')
    if var:
        
        
        ul=i.find_next_sibling('ul')
        if ul:
            for li in ul.find_all('li'):
                # text 
                text = li.get_text(strip=True)
                #url links 
                link_tag = li.find('a')["href"]
                # print(f"  - {text} ({link_tag})")

                var_list_of_dict.append(
                    {
                    "sub_header": var.text,
                    "data_point": text,
                    "url": link_tag
                    }
                )
                
                print(var_list_of_dict)
```
