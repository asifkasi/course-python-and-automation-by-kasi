```python
import requests
from bs4 import BeautifulSoup
url = ('https://www.franchising.com/')
response = requests.get(url)
# print(response.text)
response.status_code
soup = BeautifulSoup(response.text, 'html')
soup
```

```python
franc= soup.find('div',class_='brand-col1')
franc
```

```python
fran_dic ={
    'wibsite_url':'',
     'image_url':'',
     'heading':'',
     'price':'',
     'description':''
}
list = []
fran_dic['heading']=franc.find('div',class_='brand-header').text
fran_dic['wibsite_url']=franc.find('a')['href']
fran_dic['image_url']=franc.find('img')['src']
fran_dic['description']=franc.find('div',class_='brand-description').text
fran_dic['price']=franc.find('div',class_='brand-financial').text.replace('Cash Required'':',"")
list.append(fran_dic)
```
