```python
import requests
from bs4 import BeautifulSoup
url = ('https://www.yello.ae/location/dubai')
response = requests.get(url)
# print(response.text)
response.status_code
soup = BeautifulSoup(response.text, 'html')
soup
```

```python
yall= soup.find_all('div',class_='company with_img g_3')
yall
```

```python
len(yall)
```

```python
list = []
for i in yall:
    dict={
    'company name':'',
    "Establishment year":'',
    'location':'',
    'main_link':'',
    "profile_link":'',
    'Contact number':''
 }

    dict['profile_link']=i.find('a')['href']
    dict['company name']=i.find('h3').text

    dict['Contact number']=i.find('div',class_='s').text
    dict['Establishment year']=i.find_all('div',class_='s')[1].text
    dict['location']=i.find('div',class_='address').text
    list.append(dict)
```

```python
list
```

```python
import pandas as pd 
df = pd.DataFrame(list)
df
```

```python
df.to_csv('assigment no1.csv',index=False)
df
```
