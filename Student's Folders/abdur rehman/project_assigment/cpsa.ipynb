```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import time
from selenium import webdriver

driver = webdriver.Edge()

url   = "https://search.cpsa.ca/"
driver.get(url) # open the website
# driver.find_element('xpath',"//input[@id='MainContent_physicianSearchView_txtFirstName']").click()
# time.sleep(2)
# driver.find_element('xpath',"//input[@id='MainContent_physicianSearchView_txtFirstName']").send_keys("a")
# time.sleep(2)
# driver.find_element('xpath',"//input[@id='MainContent_physicianSearchView_btnSearch']").click()
# time.sleep(2)
```

```python


soup = BeautifulSoup(driver.page_source,('html'))
soup.find('div',id='MainContent_physicianSearchView_ResultsPanel')
table= soup.find('table',id='MainContent_physicianSearchView_gvResults')
trs=table.find_all('tr')
len(trs)


# Initialize an empty list to hold the data
data = []

# Find all the rows within the <tr> tags that contain physician info
rows = trs

# Loop through each row and extract the relevant information
for row in rows:
    # columns = row.find_all('td')
    name = row.find("td",class_ = "status4")
    location = row.find("td",class_ = "status4")
    practice_discipline = row.find("td",class_ = "status4")
    gender = row.find("td",class_ = "status4")
    languages = row.find("td",class_ = "status4")

    # Append the data to the list
    data.append({
        'Name': name,
        'Location': location,
        'Practice Discipline': practice_discipline,
        'Gender': gender,
        'Languages': languages
    })

# Create a DataFrame from the list of data
df = pd.DataFrame(data)

# Save the DataFrame to a CSV file
df.to_csv('physicians_data.csv', index=False)

# Print the DataFrame (optional)
print(df)
```
