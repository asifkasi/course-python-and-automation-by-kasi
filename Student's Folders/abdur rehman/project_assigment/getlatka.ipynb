```python
import pandas as pd
import time
from selenium import webdriver
from bs4 import BeautifulSoup

# WebDriver initialize karo
driver = webdriver.Chrome()

# URL jahan se data scrape karna hai
url = "https://getlatka.com/saas-companies?page=1"
driver.get(url)

# Page load hone ka intezaar karo
time.sleep(5)

# BeautifulSoup ke through page source ko parse karo
soup = BeautifulSoup(driver.page_source, "html.parser")

# Sabhi rows ko find karo (saare tr elements)
trs = soup.find_all("tr")

# Data store karne ke liye ek empty list
list_data = []

# Har row ko loop karke data extract karna
for tr in trs:
    # Har row mein saare td elements find karo
    tds = tr.find_all("td")
    
    # Agar row mein 2 se kam td elements hain, toh wo ya header ho sakta hai ya empty row ho sakti hai, skip karo
    if len(tds) < 2:
        continue

    dic = {}
    
    # Company name aur social media links ko extract karo
    links = tds[1].find_all("a")
    if links:
        dic["name"] = links[0].text  # Pehli link se company ka naam le lo
        for a in links[1:]:  # Baaki links ko loop karke extract karo
            key = a.get("aria-label", "")  # aria-label se key milti hai
            value = a.get("href", "")  # href se link milta hai
            if key and value:  # Agar key aur value dono hain, toh dictionary mein add karo
                dic[key] = value

    # Company location aur founder ki information extract karo
    dic["company location"] = tds[-2].text.strip()  # Last se second td mein location hoti hai
    dic["company founder"] = tds[-6].text.strip()  # Last se sixth td mein founder ka naam hota hai

    # Har company ka data list mein add karna
    list_data.append(dic)

# Data ko pandas DataFrame mein convert karo (optional)
df = pd.DataFrame(list_data)

# Data ko print karo
df
```

```python
df.to_csv('gelatka no1.csv',index=False)
df
```
