# 2.1 Extract All 250 movies details and save into csv.

```python
import requests
from bs4 import BeautifulSoup


header={
    "User-Agent": "Baiduspider"
}

response = requests.get('https://www.imdb.com/chart/top/',headers=header)
response.status_code
soup = BeautifulSoup(response.text, 'html.parser')
print(soup.prettify())
```

```python
var = soup.find("div",class_='ipc-metadata-list ipc-metadata-list--dividers-none sc-d24d5d37-0 hDHQeM alt_item_colors compact-list-view ipc-metadata-list--base')
# print(var)
var1 = soup.find_all("li",class_='ipc-metadata-list-summary-item')
print(var1)
# len(var1)
```

```python
list = []
for i in var1:
        dict={
            'movie name': '',
            'movie index': '',
            'movie url': '',
            'rating': '',
            'year':'',
            'movie hours': '',
            'movie vote': '',

        }
        dict['movie name']= i.find('h3').text.strip().split(". ")[1]
        dict['movie index']=i.find('h3').text.strip().split(". ")[0]
        dict['movie url'] =i.find('a')['href']
        dict['year']=i.find_all("span",class_= "sc-caa65599-7 eeMIpC cli-title-metadata-item")[0].text.strip()
        dict['movie hours']=i.find_all("span",class_= "sc-caa65599-7 eeMIpC cli-title-metadata-item")[1].text.strip()
        dict['rating']=i.find("span",class_= "ipc-rating-star--rating").text.strip()
        dict['movie vote']=i.find("span",class_= "ipc-rating-star--voteCount").text.strip()[1:-1]
        list.append(dict.copy())
```

```python
import pandas as pd 
list.append(dict)
df = pd.DataFrame(list)
df.to_csv("imdb q2.csv",index=False)
```
