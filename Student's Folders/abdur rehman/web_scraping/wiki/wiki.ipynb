```python
# Method 1: Saving data into list
import requests
from bs4 import BeautifulSoup
header={
    "User-Agent": "IsraBot"
}

r=requests.get("https://en.wikipedia.org/wiki/Python",headers=header)
soup=BeautifulSoup(r.text,'html.parser')

sub_headings=soup.find_all('div',class_='mw-heading mw-heading2')


var_list_of_dict = []

for i in sub_headings:
    var=i.find('h2')
    if var:
        # print(var.c+":")
        var_dict_data = {
            "sub_header": "",
            "data_point": "",
            "url": ""
        }
        var_dict_data["sub_header"] = var.text

        ul=i.find_next_sibling('ul')
        if ul:
            for li in ul.find_all('li'):
                # text 
                text = li.get_text(strip=True)
                var_dict_data["data_point"] = text
                #url links 
                link_tag = li.find('a')["href"]
                var_dict_data["url"] = link_tag
                # print(f"  - {text} ({link_tag})")

                var_list_of_dict.append(var_dict_data.copy())
                
                print(var_list_of_dict)
```

```python
# Method 2: Saving data into list

import requests
from bs4 import BeautifulSoup
header={
    "User-Agent": "IsraBot"
}

r=requests.get("https://en.wikipedia.org/wiki/Python",headers=header)
soup=BeautifulSoup(r.text,'html.parser')

sub_headings=soup.find_all('div',class_='mw-heading mw-heading2')


var_list_of_dict = []

for i in sub_headings:
    var=i.find('h2')
    if var:
        
        
        ul=i.find_next_sibling('ul')
        if ul:
            for li in ul.find_all('li'):
                # text 
                text = li.get_text(strip=True)
                #url links 
                link_tag = li.find('a')["href"]
                # print(f"  - {text} ({link_tag})")

                var_list_of_dict.append(
                    {
                    "sub_header": var.text,
                    "data_point": text,
                    "url": link_tag
                    }
                )
                
                print(var_list_of_dict)
```

```python
import requests
from bs4 import BeautifulSoup
import os

# Wikipedia page URL
url = "https://en.wikipedia.org/wiki/Peter_Jeffrey_(RAAF_officer)"

# Headers
headers = {"User-Agent": "IsraBot"}

# Fetch the page
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# Create folder for saving images
os.makedirs("downloaded_images", exist_ok=True)

# Find all image links inside 'a' tags with class 'image'
image_links = soup.select("a.image")

count = 1
for link in image_links:
    # Get the image page URL
    img_page_url = "https://en.wikipedia.org" + link.get("href")
    img_page_resp = requests.get(img_page_url, headers=headers)
    img_page_soup = BeautifulSoup(img_page_resp.text, "html.parser")

    # The full-size image is inside 'a' tag under div.fullImageLink
    full_img_tag = img_page_soup.select_one("div.fullImageLink a")
    if full_img_tag:
        img_url = full_img_tag.get("href")
        # Add https if URL starts with //
        if img_url.startswith("//"):
            img_url = "https:" + img_url
        # Get the file extension
        ext = img_url.split(".")[-1].split("?")[0]  # remove URL parameters
        file_name = f"downloaded_images/image_{count}.{ext}"

        try:
            print("Downloading:", img_url)
            img_data = requests.get(img_url, headers=headers).content
            with open(file_name, "wb") as f:
                f.write(img_data)
            count += 1
        except Exception as e:
            print("Failed to download", img_url, ":", e)

print(f"Downloaded {count-1} images.")
```

```python
import os
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

page_url = "https://en.wikipedia.org/wiki/Peter_Jeffrey_(RAAF_officer)"
save_dir = "images"
os.makedirs(save_dir, exist_ok=True)

headers =     {"User-Agent": "IsraBot"}
resp = requests.get(page_url, headers=headers)
resp.raise_for_status()

soup = BeautifulSoup(resp.text, "html.parser")
img_tags = soup.find_all("img")

count = 0
for img in img_tags:
    if count >= 5:
        break
    src = img.get("src")
    if not src:
        continue
    src_lower = src.lower()
    # skip obvious icons or non-jpg
    if any(x in src_lower for x in ["sprite", "logo", "icon", "thumbicon", "favicon"]) or not src_lower.endswith(".jpg"):
        continue
    # Wikipedia thumb to original
    if "/thumb/" in src:
        parts = src.split("/thumb/")
        src = parts[0] + "/" + parts[1].rsplit("/", 1)[0]
    img_url = urljoin(page_url, src)
    filename = f"img_{count+1:03d}.jpg"
    path = os.path.join(save_dir, filename)
    try:
        r = requests.get(img_url, headers=headers, timeout=10)
        r.raise_for_status()
        with open(path, "wb") as f:
            f.write(r.content)
        print("Saved:", path)
        count += 1
    except Exception as e:
        print("Failed:", img_url, e)

print("Done. Total saved:", count)
```
