## 3. Download pdfs from https://www.nbp.com.pk/RATESHEET/index.aspx

```python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://www.nbp.com.pk/RATESHEET/index.aspx")
soup = BeautifulSoup(response.text,('html'))
soup
```

```python
table =soup.find_all('tr')
len(table)
```

```python
table= soup.find('table',id='ctl00_ContentPlaceHolder1_GridView1' )
```

```python
row=table.find_all('tr')
```

```python
dict={
        'name_pdf':'',
        'pdf':'',
}
```

```python
list= []
for i in row:
    dict={
        'name_pdf':'',
        'url':'',
    }
    dict['name_pdf']= i.find('td').find('span').text
    dict['url']="https://www.nbp.com.pk/"+ i.find('a')['href'].replace('../','')
    list.append(dict)
```

```python
list
```

```python
import pandas as pd 
df = pd.DataFrame(list)
df
df.to_csv('NBP q3.csv',index=False)
df
```

```python
# for i in range(len(df)):
for i in range(5):
    url = df['url'][i]
    name_pdf_r = requests.get(url)
    name_pdf_r.status_code
    file_name=df["name_pdf"][i]+".pdf"
    with open(file_name,"wb") as f:
        f.write(name_pdf_r.content)
        f.close()
```
