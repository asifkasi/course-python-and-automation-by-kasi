```python
import requests

url = "https://catalog.data.gov/api/3/action/package_search"
params = {"sort": "metadata_created desc", "rows": 1}

response = requests.get(url, params=params)
data = response.json()

dataset = data["result"]["results"][0]

print("Most recently added dataset:")
print("Title:", dataset["title"])
print("Date added:", dataset["metadata_created"])
```

```python
import requests
from bs4 import BeautifulSoup
url = "https://catalog.data.gov/dataset"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

dataset = soup.find("h3", class_="dataset-heading")

title = dataset.text.strip()
link = "https://catalog.data.gov" + dataset.find("a")["href"]
print("Most recently added dataset on Data.gov:")
print("Title:", title)
print("Link:", link)
```
