# 	1.1  all extract first product of a page. (product_name,product_index,product_desc,product_url,img_url)


```python
import requests
from bs4 import BeautifulSoup

url = 'https://webscraper.io/test-sites'
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

```

```python
data_list = []  

var_all = soup.find_all("div", class_="row test-site")
for div in var_all:
    var_dec = {
         'product name': '',
         'product description': '',
         'product-link': '',
         'image url': ''
    }
    var_dec['product name'] = div.find('h2').text.strip()
    var_dec['product description'] = div.find('p').text.strip().replace('\n',' ').replace('\t','')
    var_dec['product-link'] = div.find('a')['href']
    var_dec['image url'] = div.find('img')['src']
    data_list.append(var_dec)
df = pd.DataFrame(data_list)
df
```

```python
import pandas as pd

df = pd.DataFrame(data_list)
df.to_csv('webscraper.csv',mode='a+',index=False)
```
