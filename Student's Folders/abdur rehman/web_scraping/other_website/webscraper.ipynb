```python
import requests
from bs4 import BeautifulSoup

respose = requests.get("https://webscraper.io/test-sites")
respose.status_code
soup = BeautifulSoup (respose.text,('html.parser'))
print(soup.prettify())
```

```python
link = soup.find_all('a')
# link
for l in link:
    print(l,['href'])
```

```python
header = soup.find_all(['h1','h2','h3','h4','h5','h6','h7'])
header
```
