## 3 https://books.toscrape.com/

```python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://books.toscrape.com/")
response.status_code 
soup= BeautifulSoup(response.text,('html.parser'))
soup
```

```python
header = soup.find_all(['h1','h2','h3','h4','h5','h6','h7'])
header
```

```python
link = soup.find_all('a')
link

```

```python
for i in link:
    print(i['href'])
```

```python
var_third_div = soup.find("li",class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")
print(var_third_div)
# var_list_of_div = soup.find_all("div",class_="container-fluid page")
```

```python
list_pro = []
var_dict = {
    "product_name":"",
    'product_index':'',
    "product_price":"",
    "product_url":"",
    "img_url":"",
    'rating':'',

}
```

```python
var_dict['product_name']=var_third_div.find("h3").text.strip()
var_dict['product_url']=var_third_div.find("a")['href']
var_dict['img_url']=var_third_div.find("img")["src"]
var_dict['product_price']=var_third_div.find("p",class_='price_color').text.strip()
var_dict['rating']=var_third_div.find("p",class_='star-rating Three').text
list_pro.append(var_dict)
```

```python
var_dict
```

```python
import pandas as pd
df = pd.DataFrame(list_pro)
df.to_csv("book_toscrape q8_1.csv", index=False)
df
```

## 3.9 Extract all product details from first page to fifth page & save into single csv file.

```python
import requests
from bs4 import BeautifulSoup

# url=("https://books.toscrape.com/catalogue/?page=")
url = "https://books.toscrape.com/catalogue/page-"
```

```python
list = []
for i in range(1,6):
    url_b = url+str(i)+'.html'
    response = requests.get(url_b)
    print(response.status_code)
    soup = BeautifulSoup(response.text,('html'))
    var_third_div = soup.find_all("li",class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")
    # list = []
    for div in var_third_div:
        var_dict = {
        "product_name":"",
        'product_index':'',
        "product_price":"",
        "product_url":"",
        "img_url":"",
        'rating':'',

        }
        var_dict['product_name']=div.find("h3").text.strip()
        var_dict['product_url']=div.find("a")['href']
        var_dict['img_url']=div.find("img")["src"]
        var_dict['product_price']=div.find("p",class_='price_color').text.strip()
        # var_dict['rating']=div.find("p",class_='star-rating Three')
        print(var_dict)
        print()
        list.append(var_dict)
    
```

```python
import pandas as pd
df = pd.DataFrame(list)
df.to_csv("book_toscrape q8_9.csv", index=False)
df
```

```python
var_fast = soup.find('article',class_='product_page')
var_fast
```

```python
var_dict['product_name']=var_fast.find('h1').text
var_dict['product_url']=var_fast.find("a")['href']
var_dict['img_url']=var_fast.find("img")["src"]
var_dict['product_price']=var_fast.find("p",class_='price_color').text.strip()
var_dict['rating']=var_fast.find("p",class_='star-rating Three').text
var_dict['UPC']=var_fast.find('tr').text.strip()
list_product.append(var_dict)
```
