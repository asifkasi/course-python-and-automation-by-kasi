## 8.1 Extract first product details save into csv file. e.g# A Light in the ...
## https://books.toscrape.com/

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

main_url = ('https://books.toscrape.com')
response = requests.get(main_url)
response.status_code
soup = BeautifulSoup(response.text,('html'))
print(soup.prettify())
```

```python
var_fast = soup.find('li',class_='col-xs-6 col-sm-4 col-md-3 col-lg-3')
var_fast
```

```python
len(var_fast)
```

```python
var_fast.find('p',class_='instock availability').text.strip()
```

```python
list_product = []
var_dict = {
    "Product_name":'',
    'product_url':'',
    "Product_Description":'',
    "UPC":'',
   'Tax':'',
    'product_reviews_star_no':'',
    'product_price':'',
    "product_availabililty":'',
}
var_dict['Product_name']=var_fast.find('h3').text
var_dict['product_url']=var_fast.find("a")['href']
var_dict['img_url']=var_fast.find("img")["src"]
var_dict['product_price']=var_fast.find("p",class_='price_color').text.strip()
var_dict['product_reviews_star_no']=var_fast.find("p",class_ = "star-rating").attrs["class"][1]
var_dict['product_availabililty']=var_fast.find('p',class_='instock availability').text.strip()
list_product.append(var_dict)
var_dict


act_url = urljoin(main_url, var_dict["product_url"])
resp = requests.get(act_url)
soup_nw = BeautifulSoup(resp.text, "html.parser")
var_dict["Product_Description"] =  soup_nw.find("div",id = "product_description").find_next_sibling().text.strip()


for i in soup_nw.find_all("tr"):
    var_dict[i.find("th").text] = i.find("td").text

var_dict["Product_name"] = soup_nw.find("h1").text
```

```python
var_dict
```

```python
import pandas as pd 
df = pd.DataFrame(list_product)
df
df.to_csv('toscrape q8_1.csv',index=False)
df
```

```python
list_product = []
for div in var_fast[3:20:4]:
    var_dict = {
        "Product_name":'',
        'product_url':'',
        "Product_Description":'',
        "UPC":'',
        'Tax':'',
        'product_reviews_star_no':'',
        'product_price':'',
        "product_availabililty":'',
    }
    var_dict['Product_name']=div.find('h3').text
    var_dict['product_url']=div.find("a")['href']
    var_dict['img_url']=div.find("img")["src"]
    var_dict['product_price']=div.find("p",class_='price_color').text.strip()
    var_dict['product_reviews_star_no']=div.find("p",class_ = "star-rating").attrs["class"][1]
    var_dict['product_availabililty']=div.find('p',class_='instock availability').text.strip()
    list_product.append(var_dict)
    var_dict


    act_url = urljoin(main_url, var_dict["product_url"])
    resp = requests.get(act_url)
    soup_nw = BeautifulSoup(resp.text, "html.parser")
    var_dict["Product_Description"] =  soup_nw.find("div",id = "product_description").find_next_sibling().text.strip()


    for i in soup_nw.find_all("tr"):
        var_dict[i.find("th").text] = i.find("td").text

    var_dict["Product_name"] = soup_nw.find("h1").text
```

```python
import pandas as pd 
df = pd.DataFrame(list_product)
df
df.to_csv('toscrape q8_7.csv',index=False)
df
```
