```python
import requests
from bs4 import BeautifulSoup

def scrape_first_product(url):
    """
    Extract first product information from books.toscrape.com
    Returns: product_name, product_index, product_desc, product_url, img_url
    """
    
    # Send request to the website
    response = requests.get(url)
    response.raise_for_status()
    
    # Parse HTML
    soup = BeautifulSoup(response.content, 'html.parser')
    
    # Find first product
    first_product = soup.find('article', class_='product_pod')
    
    if not first_product:
        return None
    
    # Extract product name
    product_name = first_product.find('h3').find('a')['title']
    
    # Product index (first product)
    product_index = 1
    
    # Extract product URL
    product_link = first_product.find('h3').find('a')['href']
    product_url = f"https://books.toscrape.com/catalogue/{product_link}"
    
    # Extract image URL
    img_tag = first_product.find('img')
    img_url = f"https://books.toscrape.com/{img_tag['src']}"
    
    # Get product description (need to visit product page)
    product_response = requests.get(product_url)
    product_soup = BeautifulSoup(product_response.content, 'html.parser')
    
    # Extract description
    product_desc_tags = product_soup.find_all('meta', {'name': 'description'})
    product_desc = product_desc_tags[0]['content'].strip() if product_desc_tags else "No description available"

    return {
        'product_name': product_name,
        'product_index': product_index,
        'product_desc': product_desc,
        'product_url': product_url,
        'img_url': img_url
    }


# Main execution
if __name__ == "__main__":
    url = "https://books.toscrape.com/"
    
    print("Scraping first product from Books to Scrape...\n")
    
    result = scrape_first_product(url)
    
    if result:
        print(f"Product Name: {result['product_name']}")
        print(f"Product Index: {result['product_index']}")
        print(f"Product Description: {result['product_desc']}")
        print(f"Product URL: {result['product_url']}")
        print(f"Image URL: {result['img_url']}")
    else:
        print("Could not find any product on the page.")
```
