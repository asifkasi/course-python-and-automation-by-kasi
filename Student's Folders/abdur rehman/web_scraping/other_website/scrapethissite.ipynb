## 2 https://www.scrapethissite.com/pages/

```python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://www.scrapethissite.com/pages/")
response.status_code
soup = BeautifulSoup(response.text,('html.parser'))
soup
```

```python
link = soup.find_all('a')
link

```

```python
for i in link:
    print(i['href'])
```

```python
header = soup.find_all(['h1','h2','h3','h4','h5','h6','h7'])
header
```

```python
var_all = soup.find_all('h3')
var_all
```

```python
var_all = soup.find_all('div',class_='page')
var_all
# sec_dec_list = []
```

```python
data = []
var_all = soup.find_all('div',class_='page')
for div in var_all:
    var_dict = {
    "product_name":"",
    "product_desc":"",
    "product_url":"",
    # "img_url":""

    }
    var_dict["product_name"] = div.find("h3").text.strip().replace("\n"," ").replace("\t","")
    var_dict["product_desc"] = div.find("p").text.strip().replace("\n"," ").replace("\t","")
    var_dict["product_url"] = div.find("a")["href"]
    # var_dict
#     var_dict["img_url"] = div.find("img")["src"]
    # print(var_dict)
    # print()
    data.append(var_dict.copy())
import pandas as pd    
df = pd.DataFrame(data)
df
```
