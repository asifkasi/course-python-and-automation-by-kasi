```python
import requests
from bs4 import BeautifulSoup    
url = "https://www.tutorialsfreak.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
images = soup.find_all('img')
for image in images:
    print(image ,stripe = True)
```
