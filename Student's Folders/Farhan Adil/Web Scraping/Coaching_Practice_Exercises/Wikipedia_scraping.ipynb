# 1. Write a Python program to extract and display all the header tags from en.wikipedia.org/wiki/Main_Page

```python
import requests # we use requests to fetch the content of the webpage
from bs4 import BeautifulSoup  # we use BeautifulSoup to parse the HTML content
url = "https://en.wikipedia.org/wiki/Main_Page" # URL of the webpage we want to scrape
# we use header to bypass Blocking of IP by website
headers = {
    "User-Agent": "Orthogaffe"
}
response = requests.get(url ,headers=headers) #fetching the content of the webage
soup = BeautifulSoup(response.text , 'html.parser') # parsing the HTML content
```

```python
# EXtracting all headings from h1 to h7
headings = soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6' , 'h7']) # finding all heading tags
# printing the extracted headings
for heading in headings: # iterating through each heading
    print(heading.text)
```

## 2. Write a Python program to extract and display all the image links from en.wikipedia.org/wiki/Peter_Jeffrey_(RAAF_officer).

```python
import requests # we use requests to fetch the content of the webpage
from bs4 import BeautifulSoup 
url = "https://en.wikipedia.org/wiki/Peter_Jeffrey_(RAAF_officer)" # url of the webpage we want to scrape
# we use header to bypass Blocking of IP by website
headers = {
    "User-Agent": "Orthogaffe"
}

response = requests.get(url ,headers=headers) #fetching the content of the webage
soup = BeautifulSoup(response.text , 'html.parser') # parsing the HTML content

# Extracting all image URLs from the webpage
images = soup.find_all('img')
for image in images: # iterating through each image tag
    print(image['src']) # printing the src attribute of each image tag
   
```

## 3. Write a Python program to that retrieves an arbitary Wikipedia page of "Python" and creates a list of links on that page.

```python
import requests # we use request library
from bs4 import BeautifulSoup # we use beautifulsoup 

url = "https://en.wikipedia.org/wiki/Python_(disambiguation)" # url of a webpage
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"} #we use bot to scrape website

response = requests.get(url, headers=headers) # response store html text of a wbpage
soup = BeautifulSoup(response.text, "html.parser") # we parse the html content

ul_par = soup.find('div', class_="mw-content-ltr mw-parser-output") # we use class to scrape of this where is Ul
Uls = ul_par.find_all( 'ul' ) # we find all ul tag in this class

links = [] # empty list to store links

for tag in Uls: # iterating through each ul tag
    for a in tag.find_all('a', href=True): # finding all anchor tags with href attribute
        full_url = "https://en.wikipedia.org" + a['href'] # constructing full URL
        links.append(full_url) # append full_url to link (list)
        print(full_url) # printing the full URL
```

## 4. Write a Python program to list all language names and number of related articles in the order they appear in wikipedia.org.

```python

import requests # we use request library
from bs4 import BeautifulSoup # we use beautifulsoup

# Step 1: Target the main Wikipedia homepage
url = "https://www.wikipedia.org/" # url of a webpage
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}

response = requests.get(url, headers=headers) # response store html text of a wbpage
soup = BeautifulSoup(response.text, "html.parser") # we parse the html content

# Step 2: Find all language boxes
languages = soup.find_all("a", class_="link-box") # finding all anchor tags with class 'link-box'

# Step 3: Extract and display names and article counts
for lang in languages: # iterating through each language box
    name = lang.find("strong").text.strip() 
    articles = lang.find("small").text.strip() 
    print(f"{name} : {articles}") # print `name` and ` articles`
```
