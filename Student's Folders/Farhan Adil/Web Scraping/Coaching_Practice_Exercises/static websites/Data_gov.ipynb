## 1. Write a Python program to get the number of datasets currently listed on data.gov.

```python
import requests
from bs4 import BeautifulSoup
response = requests.get("https://data.gov/")
soup = BeautifulSoup(response.text, "html.parser")

data = soup.find("h4", class_="text-uppercase")

print(data.text.strip())
```

## 2. Write a Python program to display the name of the most recently added dataset on data.gov

```python
import requests
from bs4 import BeautifulSoup

# Step 1: Go to the dataset catalog page (sorted by newest)
url = "https://catalog.data.gov/dataset?q=&sort=metadata_created+desc"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# Step 2: Find the first dataset title (the most recent one)
dataset = soup.find("h3", class_="dataset-heading")

if dataset and dataset.a:
    latest_dataset = dataset.a.text.strip()
    print("Most recently added dataset on Data.gov:")
    print(latest_dataset)
else:
    print("Could not find the dataset title.")
    
```
