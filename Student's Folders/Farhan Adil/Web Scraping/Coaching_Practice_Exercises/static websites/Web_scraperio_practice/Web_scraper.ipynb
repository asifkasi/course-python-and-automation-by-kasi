## 4.1 Extract first Table of a page & save into another csv file.

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
r= requests.get("https://webscraper.io/test-sites/tables")
r.status_code
```

```python
soup = BeautifulSoup(r.text , "html.parser")
soup.prettify()
```

```python
tb = soup.find("table" , class_ = "table table-bordered")
tb
```

```python
trs = tb.find_all("tr")
trs
```

```python
dl = []
for td in trs:
    dr = {
    "First Name" : "" ,
    "Last Name" : "" ,
    "Username" : ""
}
    cols = td.find_all("td")
    if len(cols) >= 3:
         dr["First Name"] = cols[1].text.strip()
         dr["Last Name"] = cols[2].text.strip()
         dr["Username"] = cols[3].text.strip()
         dl.append(dr)
dl
```

```python
df = pd.DataFrame(dl)
df.to_csv("webscraper_q4.1.csv" , index = False)
```

```python
dl = []
for product in products:
    
    data = {
        "product_name" : "",
        "product_link" : "",
        "product_img" : "",
        "product_price" : "",
        "product_details" : "",
        "product_reviews_number" : "",
    # "product_reviews_star_number" : ""
    }
    
    data["product_link"] = product.find("a")["href"] # product url

    base_url = "https://webscraper.io"
    act_url = base_url + data["product_link"]
    req = requests.get(act_url)
    soup_nw = BeautifulSoup(req.text , "html.parser")
    data["product_name"] = soup_nw.find("h4" , class_ = "title card-title" ).text

    data["product_img"] = product.find("img")["src"].strip() # img 
    data["product_price"] = product.find("span").text # price
    data["product_details"] = product.find("p" , class_ = "description card-text").text
    data["product_reviews_number"] = product.find("p" , class_ = "review-count float-end").text.strip()
    dl.append(data)

dl
```

```python
df = pd.DataFrame(dl)
df
```

```python
df.to_csv("web_scraper_qa5.3.csv")
```
