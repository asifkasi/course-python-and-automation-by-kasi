```python
import pandas as pd
data = [] # create an empty list to store book data

for book in header:
    title = book.find("h3").text # extract the book title
    link = book.find("a")["href"] # extract the book link
    price = book.find("p", class_ = "price_color").text # extract the book price

    book_data = { # create a dictionary to store book details
    "Title" : "",
    "Link" : "",
    "Price" : ""
}
    book_data["Title"] = title # assign title to the dictionary
    book_data["Link"] = link  # assign link to the dictionary
    book_data["Price"] = price # assign price to the dictionaryv
    data.append(book_data) # append the dictionary to the list
    df = pd.DataFrame(data) # create a DataFrame from the list of dictionaries
    df

```
