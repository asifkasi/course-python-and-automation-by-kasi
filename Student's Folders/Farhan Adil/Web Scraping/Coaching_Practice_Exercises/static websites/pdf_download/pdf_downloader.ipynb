## a3. Download pdfs from https://www.nbp.com.pk/RATESHEET/index.aspx

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin 

url = "https://www.nbp.com.pk/RATESHEET/index.aspx"

# Load page
page = requests.get(url)
page.status_code
```

```python

soup = BeautifulSoup(page.text, "html.parser")

count = 0

for a in soup.find_all("a", href=True):
    href = a["href"]

    if href.lower().endswith(".pdf"):
        pdf_url = urljoin(url, href)  # FIX

        file_name = pdf_url.split("/")[-1]
        print("Downloading:", file_name)

        pdf = requests.get(pdf_url)
        with open(file_name, "wb") as f:
            f.write(pdf.content)

        count += 1
        if count == 5:
            break

print("Downloaded first 5 PDFs.")
```

### Another method

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://www.nbp.com.pk/RATESHEET/index.aspx"

# Load page
page = requests.get(url)
page.status_code
```

```python
soup = BeautifulSoup(page.text ,"html.parser")
table = soup.find("table" , id = "ctl00_ContentPlaceHolder1_GridView1")
len(table)
```

```python
tr = table.find_all("tr")
tr
```

```python
li = []
b_url = "https://www.nbp.com.pk/" 
for t in tr:

    data = {"File Name" : "" ,
        "File Url" : ""}
    data["File Name"] = (t.find("span")).text
    data["File Url"] = ( "https://www.nbp.com.pk/" + t.find("a") ["href"].replace("../",""))
    li.append(data)
li
```

```python
df = pd.DataFrame(li)
df
```

```python
df["File Url"][0]
df["File Name"][0]
```

```python
for i in range(5):
    url_pdf = df["File Url"][i]
    pdf_r = requests.get(url_pdf)
    pdf_r.status_code

    file_name = df["File Name"][i]  + ".pdf"

    with open(file_name , "wb") as f:
        f.write(pdf_r.content)

        

```
