### a1.2 Takes Pakistani city name as input from user, and find {temprature, wind, wind gusts, air quality}. Note: Make a list of city, convert input into lowercase and check city name is correct or not.  

```python
import requests
from bs4 import BeautifulSoup
header = {
    "User-Agent" : "Googlebot-News"
}

cities = ["karachi", "lahore", "islamabad", "peshawar", "quetta", "faisalabad", "multan", "hyderabad"]

user_city = input("enter your city").strip().lower()
if user_city != cities:
    print("enter valid city")
search_url = f"https://www.accuweather.com/en/search-locations?query={user_city}"
r = requests.get(search_url , headers = header)
soup = BeautifulSoup(r.text , "html.parser")

key_url = soup.find("div" , class_ = "locations-list content-module").find("a")["href"]

b_url = "https://www.accuweather.com"

main_url = b_url + key_url
r1 = requests.get(main_url , headers = header)
soup1 = BeautifulSoup(r1.text , "html.parser")

data = soup1.find("div" , class_ = "page-content content-module")

li = []
da = {
    "temperature" : "" ,
    "wind" : "" ,
    "wind gusts" : "" ,
    "air quality" : ""
}
da["temperature"] = data.find("div" , class_= "temp").text
da["wind"] = data.find_all("div" , class_ ="spaced-content detail")[0].find("span" , class_= "value").text
da["wind gusts"] = data.find_all("div" , class_ ="spaced-content detail")[1].find("span" , class_= "value").text
da["air quality"] = data.find_all("div" , class_ ="spaced-content detail")[2].find("span" , class_= "value").text
li.append(da)
li




```
