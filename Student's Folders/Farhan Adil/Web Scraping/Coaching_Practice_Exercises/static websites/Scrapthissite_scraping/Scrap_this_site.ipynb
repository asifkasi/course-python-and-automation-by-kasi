##  2. https://www.scrapethissite.com/pages/ (header, paragraph, header_link)
	## 2.1 Extract first paragraph details of a page & save into csv file.
	## 2.2 Extract all paragraphs details of a page & save into csv file.

```python
import requests # requests Library
from bs4 import BeautifulSoup
import pandas as pd
url = "https://www.scrapethissite.com/pages/" # store Web page url in variable
r = requests.get(url)
r.status_code
```

```python
soup = BeautifulSoup(r.text, "html.parser") # parsing Webpage
soup.prettify()
```

```python
body = soup.find("div" , class_= "col-md-6 col-md-offset-3") # storing div
body.prettify()
```

```python
body.find_all("h3") # finding H3
```

```python
body.find_all("p") # finding paragraphs
```

```python
body.find_all("a" , href = True) # finding Links
```

```python
dis=body.find_all("div",class_="page") 
dis
```

```python
dis[0] # indexing
```

```python
data = []
for div in dis:  # for loop
    ul = {
        "header" : "" ,
        "para" : "" ,
        "link" : "" 
    }
    ul["header"] = div.find("h3").text.strip()
    ul["para"] = div.find ("p").text.strip()
    ul["link"] = div.find ("a")["href"]
    data.append(ul)
df = pd.DataFrame(data)
df



   

```

```python
print(dis[0].find("h3").text.strip())
print(dis[0].find("p").text.strip())
dis[0].find("a")["href"]
```

## 3. https://www.scrapethissite.com/pages/forms/ (Team Name, Year, Wins, Losses, OT Losses, Win %, Goals For (GF), Goals Against (GA), + / -)

## 3.1 Extract Table of first page & save into csv file.

```python

import requests
from bs4 import BeautifulSoup
r = requests.get("https://www.scrapethissite.com/pages/forms/")
r.status_code
```

```python
soup = BeautifulSoup(r.text , "html.parser")
soup.prettify
```

```python
data = {
    "Team_name" : "",
    "Year" : "",
    "Wins" : "" ,
    "Losses" : "",
    "Win_%" : "" , 
    "Goal_for" : "",
    "Goal_Against" : "",
    "+/-" : ""
}
```

```python
tr = soup.find("table" , class_ = "table")
```

```python
td = tr.find_all("tr" , class_ = "team")
```

```python
tr.find("td" , class_ = "name").text.strip() # Name
```

```python
tr.find("td" , class_ = "year").text.strip()
```

```python
dl = []
for i in td:
    data = {
    "Team_name" : "",
    "Year" : "",
    "Wins" : "" ,
    "Losses" : "",
    "Ot_losses" : "",
    "Win_%" : "" , 
    "Goal_for" : "",
    "Goal_Against" : "",
    "+/-" : ""
}   
    data["Team_name"] = i.find("td" , class_ = "name").text.strip()
    data["Year"] = i.find("td" , class_ = "year").text.strip()
    data["Wins"] = i.find("td" , class_ = "wins").text.strip()
    data["Losses"] = i.find("td" , class_ = "losses").text.strip()
    data["Ot_losses"] = i.find("td" , class_ = "ot-losses").text.strip()
    try: 
          data["win%"] = i.find("td" , class_ = "pct text-danger").text.strip()
    except:
          data["win%"] = i.find("td" , class_ = "pct text-success").text.strip()
    data["Goal_for"] = i.find("td" , class_ = "gf").text.strip()
    data["Goal_Against"] = i.find("td" , class_ = "ga").text.strip()
    
    data["+/-"] = i.find("td" , class_ = "diff").text.strip()
    dl.append(data)
dl

```

```python
import pandas as pd
df = pd.DataFrame(dl)
df
```

```python
df.to_csv("Scrapthissite_q3.1.csv" , index = False)
```
