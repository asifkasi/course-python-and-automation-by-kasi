## https://webscraper.io/test-sites Scrap Productname, index, desc, url, img_url

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
url = "https://webscraper.io/test-sites"
response = requests.get(url)
soup = BeautifulSoup(response.content, "html.parser")
soup.prettify()
```

```python
var_dict = {
    "product_name":"",
    "product_index":"1",
    "product_desc":"",
    "product_url":"",
    "img_url":""

}
```

```python
var_fst_div = soup.find("div",class_="row test-site")
print(var_fst_div.prettify())
```

```python
var_dict["product_name"] = var_fst_div.find("h2").text.strip()
var_dict["product_desc"] = var_fst_div.find("p").text.strip().replace("\n"," ").replace("\t","")
var_dict["product_url"] = var_fst_div.find("a")["href"]
# var_dict
var_dict["img_url"] = var_fst_div.find("img")["src"]
var_dict
```

```python
var_fst_div = soup.find("div",class_="row test-site")
var_lsit_of_div = soup.find_all("div",class_="row test-site")
```

```python
list = []
for div in var_lsit_of_div:
    var_dict = {
    "product_name":"",
    # "product_index":"1",
    "product_desc":"",
    "product_url":"",
    "img_url":""

    }
    var_dict["product_name"] = div.find("h2").text.strip()
    var_dict["product_desc"] = div.find("p").text.strip().replace("\n"," ").replace("\t","")
    var_dict["product_url"] = div.find("a")["href"]
    # var_dict
    var_dict["img_url"] = div.find("img")["src"]
    print(var_dict)
    print()
    list.append(var_dict)
```
