# 	1.1 extract first product of a page. (product_name,product_index,product_desc,product_url,img_url)


```python
import requests
from bs4 import BeautifulSoup
```

```python
r = requests.get("https://webscraper.io/test-sites")
r.status_code
```

```python
soup = BeautifulSoup(r.text,"html")
print(soup.prettify())
```

```python
var_dict = {
    "product_name":"",
    "product_index":"1",
    "product_desc":"",
    "product_url":"",
    "img_url":""

}
```

```python
var_fst_div = soup.find("div",class_="row test-site")
print(var_fst_div.prettify())
# .text
```

```python
var_dict["product_name"] = var_fst_div.find("h2").text.strip()
var_dict["product_desc"] = var_fst_div.find("p").text.strip().replace("\n"," ").replace("\t","")
var_dict["product_url"] = var_fst_div.find("a")["href"]
# var_dict
var_dict["img_url"] = var_fst_div.find("img")["src"]
var_dict
```
