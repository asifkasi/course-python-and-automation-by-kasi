```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
main_url = "https://books.toscrape.com/"
r = requests.get(main_url)
r.status_code
```

```python
soup = BeautifulSoup(r.text,"html")
products = soup.find_all("li",class_ = "col-xs-6 col-sm-4 col-md-3 col-lg-3") # finding all products div
print("total products are ",len(products))
```

```python
list = []
for fst_pro in products:
    dict_pro = {
    "Product_name":"",
    "Product_url":"",
    "product_img":"",
    "product_price":"",
    "product_availabililty":""
    }
    dict_pro["Product_name"]= fst_pro.find("h3").text
    dict_pro["product_price"]= fst_pro.find("p",class_ = "price_color").text
    dict_pro["product_availabililty"]= fst_pro.find("p",class_ = "instock availability").text.strip()
    dict_pro["product_img"]= fst_pro.find("img")["src"]
    dict_pro["Product_url"]= fst_pro.find("a")["href"]
    list.append(dict_pro)

df  = pd.DataFrame(list)
df
    
```
