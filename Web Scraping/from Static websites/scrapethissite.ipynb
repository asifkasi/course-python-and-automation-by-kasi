```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url ="https://www.scrapethissite.com/pages/"

r = requests.get(url)

print(r.status_code)

soup = BeautifulSoup(r.text,"html")
# print(soup)

# divs = soup.find("div",class_="page") #it will give the first one
divs = soup.find_all("div",class_="page") #it will give all divs in list

dist = {
    "header":"",
    "para":"",
    "link":"",

}
list=[]

for div in divs:

    # print("(***********)")

    # print("h3 value: ",div.find("h3").text.strip())
    # print("p value: ",div.find("p").text.strip())
    # print("link value: ",div.find("a")["href"])
    
    dist["header"] = div.find("h3").text.strip()
    dist["para"] = div.find("p").text.strip()
    dist["link"] = div.find("a")["href"]

    list.append(dist.copy())

list

```

```python
df = pd.DataFrame(list)
df
```

```python
df.to_csv("data of scrapethissite.csv",index=False)
```

##	3.1 Extract Table of first page & save into csv file.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url ="https://www.scrapethissite.com/pages/forms/"

r = requests.get(url)

print(r.status_code)

soup = BeautifulSoup(r.text,"html")
```

```python
tbl = soup.find('table',class_ = "table")
tbl
```

```python
trs = tbl.find_all("tr")
len(trs)
```

### Method 1

```python
list_df = []
list_row  = []
for th in trs[0].find_all("th"):
    # print(th.text.strip())
    list_row.append(th.text.strip())
list_row
list_df.append(list_row)
```

```python
trs[1].find_all('td')
```

```python
for tr in trs[1:]:
    list_row= []
    for col in tr.find_all("td"):
        # print(col.text.strip(),end=" ")
        list_row.append(col.text.strip())
    # print()
    list_df.append(list_row)
df = pd.DataFrame(list_df[1:], columns=list_df[0])
df
    
```

### method 2

```python
dict = {
'Team Name' : "",
'Year' : "",
'Wins' : "",
'Losses' : "",
'OT Losses' : "",
'Win %' : "",
'Goals For (GF)' : "",
'Goals Against (GA)' : "",
'+ / -' : ""
}
```

```python
list_df = []
for tr in trs[1:]:
    cols = tr.find_all("td")
    dict["Team Name"] = cols[0].text.strip()
    dict["Year"] = cols[1].text.strip()
    dict["Wins"] = cols[2].text.strip()
    dict["Losses"] = cols[3].text.strip()
    dict["OT Losses"] = cols[4].text.strip()
    dict["Win %"] = cols[5].text.strip()
    dict["Goals For (GF)"] = cols[6].text.strip()
    dict["Goals Against (GA)"] = cols[7].text.strip()
    dict["+ / -"] = cols[8].text.strip()
    list_df.append(dict.copy())
    
list_df
```

```python
df = pd.DataFrame(list_df)
df
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
r=requests.get('https://www.scrapethissite.com/pages/forms/')
print(r.status_code)
soup=BeautifulSoup(r.text,'html.parser')
```

```python

web=soup.find_all('tr',class_='team')
list = [] 
for table in web[1:]:
    dic={
        'Team Name':'',
        'Year':'',
        'Wins':'',
        'Losses':'',
        'OT Losses':'Nill',
        'Win %':'',
        'Goals For (GF)':'',
        'Goals Against (GA)':'',
        '+ / -':'',
    }
    dic['Team Name']=table.find('td',class_='name').text.strip()
    dic['Year']=table.find('td',class_='year').text.strip()
    dic['Wins']=table.find('td',class_='wins').text.strip()
    dic['Losses']=table.find('td',class_='losses').text.strip()
    
    try:
        dic['Win %']=table.find('td',class_='pct text-success').text.strip()
    except:
        dic['Win %']=table.find('td',class_='pct text-danger').text.strip()

    dic['Goals For (GF)']=table.find('td',class_='gf').text.strip()
    dic['Goals Against (GA)']=table.find('td',class_='ga').text.strip()
    # dic['+ / -']=table.find('td',class_='diff text-success').text.strip()
    dic['+ / -']=table.find('td',class_='diff').text.strip()

    # print(dic)
    # list.append(dic.copy())
    list.append(dic)
print(list)
```

```python
df = pd.DataFrame(list)
df
```
