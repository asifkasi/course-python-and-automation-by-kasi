```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url  = "https://www.nbp.com.pk/RATESHEET/index.aspx"
r = requests.get(url)
r.status_code
```

```python
soup = BeautifulSoup(r.text,"html")
print(soup.prettify())
```

```python
tbody = soup.find("table",id = "ctl00_ContentPlaceHolder1_GridView1")
len(tbody)

div_pdfs = tbody.find_all("tr")
len(div_pdfs)
```

```python

# print(i.find("a")["href"])
```

```python
list = []
for i in div_pdfs:
    # print(i.text.strip())
    # print(i.text.strip())

    # print(i.find("span").text.strip(),end = " ")
    # print("https://www.nbp.com.pk/" + i.find("a")["href"])
    # print("https://www.nbp.com.pk/" + i.find("a")["href"].replace("../",""))
    list.append({
        "pdf_name":i.find("span").text.strip(),
        "url":"https://www.nbp.com.pk/" + i.find("a")["href"].replace("../","")

    })
    # .text.strip()
list
```

```python
df = pd.DataFrame(list)
df
```

```python
for i in range(len(df)):
# for i in range(5):
    url_pdf  = df["url"][i]
    pdf_r = requests.get(url_pdf)
    pdf_r.status_code
    file_name = "D:\Github\course_python_n_automation\Web Scraping\pdf\\"+df["pdf_name"][i]+".pdf"
    # file_name = folder_path +"\\"+ file_name + file_extension 
    # file_name = df["pdf_name"][i]+".pdf"

    with open(file_name,"wb") as f:
        f.write(pdf_r.content)
    f.close()
```
