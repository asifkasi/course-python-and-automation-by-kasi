## Method 1 (Extractacting detail of inner page via Request library)

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import requests
import time

driver = webdriver.Chrome()
url = "https://www.scrapethissite.com/pages/frames/"
driver.get(url)
driver.switch_to.frame(0)
driver.find_element("xpath",'//h3').text
soup = BeautifulSoup(driver.page_source,('html'))
frm = soup.find_all('div',class_='row')
frm
```

```python
list =[]
for i in frm[0:3]:
    base_url = 'https://www.scrapethissite.com/'
    dict = {
        'name':'',
        'url':'',
        'img_url':'',
        "detail":'',
    }
    dict ['img_url']=i.find('img')['src']
    dict ['inner_url']=base_url + i.find('a')['href']
    dict ['name']=i.find('h3').text
    
    inner_url = dict["inner_url"] # inner url
    headers = {"User-Agent": "Mozilla/5.0"}
    
    resp = requests.get(inner_url, headers=headers)

    soup_nw = BeautifulSoup(resp.text, "html.parser")
    # soup_nw
    dict["detail"]=soup_nw.find('p').text.strip()# all url details
    
    list.append(dict)
```

```python
df = pd.DataFrame(list)
df
```

## Method 2
'''
1. clicking on `Learn More` button and extracting details.
2. Clicking on `Back to all turtles` button.
3. clicking on next div (family/product) `Learn More` button and extracting details.
4. repeat`


'''

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import requests
import time

driver = webdriver.Chrome()
url = "https://www.scrapethissite.com/pages/frames/"
driver.get(url)
driver.switch_to.frame(0)
driver.find_element("xpath",'//h3').text


# soup = BeautifulSoup(driver.page_source,('html'))
frm = driver.find_elements('xpath',"//div[@class='col-md-4 turtle-family-card']")
len(frm)
```

```python
list =[]
for j in range(3):
    base_url = 'https://www.scrapethissite.com/'
    frm = driver.find_elements('xpath',"//div[@class='col-md-4 turtle-family-card']")
    i  = frm[j]
    dict = {
        'name':'',
        'url':'',
        'img_url':'',
        "detail":'',
    }
    dict ['img_url'] = i.find_element("xpath",'img').get_attribute("src")

    dict ['inner_url']=base_url + i.find_element("xpath",'a').get_attribute('href')
    dict ['name']=i.find_element("xpath",'h3').text

    # 1. clicking on `Learn More` button and extracting details.
    i.find_element("xpath","//a[@class='btn btn-default btn-xs']").click()
    time.sleep(2)

    dict["detail"]= driver.find_element("xpath","//p[@class= 'lead']").text
    list.append(dict)

    #2. Clicking on `Back to all turtles` button.
    driver.find_element("xpath","//a[@class='btn btn-default btn-xs']").click()
    time.sleep(8)
    # driver.switch_to.frame(0)
    

    
```

```python
list = []
for div in divs[:3]:
    dict = {
        "name":"",
        "url":"",
        "img_url":""
    }
    base_url = "https://www.scrapethissite.com/"
    dict["img_url"] = div.find("img")["src"]
    dict["url"] =base_url + div.find("a")["href"]
    dict["name"] =div.find("h3").text
    # dict
    list.append(dict)

df = pd.DataFrame(list)
df
```

## 	1.7 Extract Tables from year 2015 to 2010 & save into single csv file.

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import time

driver = webdriver.Chrome()
url = "https://www.scrapethissite.com/pages/ajax-javascript/"
driver.get(url)

for yr in range(2015,2009,-1):
    # driver.find_element("xpath","//element_name[@class='']")
    # driver.find_element("xpath","//element_name[@id='']")
    # driver.find_element("xpath",'//element_name[@id=""]')
    driver.find_element("xpath",f"//a[@id='{yr}']").click() # click on each year to load table data
    print(yr)
    time.sleep(3) #time delay to oad the table

    soup=BeautifulSoup(driver.page_source,"html")
    tbs= soup.find_all("table",class_= "table")[0]
    # tbs= soup.find("table",class_= "table")

    trs = tbs.find_all("tr")
    list = []
    for i in trs[1:]:
        dict = {        
        "title":i.find("td",class_ ="film-title" ).text.strip(),
        "Nomination":i.find("td",class_ ="film-nominations" ).text.strip(),
        "awards":i.find("td",class_ ="film-awards" ).text.strip()
        }
        list.append(dict.copy())
    df = pd.DataFrame(list)
    # df.to_csv("scrape_this_site_yr_"+str(yr)+".csv",index=False)
    df.to_csv(f"scrape_this_site_yr_{yr}.csv",index=False)
```
