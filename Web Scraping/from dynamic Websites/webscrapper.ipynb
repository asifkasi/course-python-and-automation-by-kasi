```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
import time

driver = webdriver.Chrome()
url = "https://webscraper.io/test-sites/e-commerce/ajax/computers/laptops"
driver.get(url)
```

```python
# method 1 
'''finding all buttons with same name, find the specific button number via condition and click it.'''
for i in range(2,6):
    for btn in driver.find_elements("xpath","//button[@class = 'btn btn-default page-link page']"):
        if btn.text == str(i):
            print("clicking on page:",btn.text)
            btn.click()
            break
```

```python
# method2: find the parent button then find all buttons in it, click via indexing accordingly 
btns  = driver.find_element("xpath","//div[@class = 'btn-group pagination justify-content-center']").find_elements("xpath","//button")
# btns[2].text #page 1 button
# btns[3].text #page 2 button
# btns[4].text #page 3 button
# btns[21].text #page 20 button
for i in range(1,5):
    btns[i+2].click()
    print("clicking on page:",btns[i+2].text)


```

```python
# method3: using sibling
'''
find the active button, and click next sibling
'''
for i in range(4):
    atv_btn = driver.find_element("xpath","//button[@class = 'btn btn-default page-link page active btn-primary']")
    atv_btn.find_element("xpath", "following-sibling::button").click()
    print("clicking on page:",atv_btn.find_element("xpath", "following-sibling::button").text)


```

```python
# method4:using contains function within find_elements function
'''
this will find all buttons included active button, indexing will be start from 0 to 19
'''
btns = driver.find_elements("xpath", "//button[contains(@class, 'btn btn-default page-link page')]")
len(btns)
# btns[0].text #page 1 button
# btns[1].text #page 2 button
# btns[2].text #page 3 button
# btns[19].text #page 20 button
for i in range(1,5):
    print("clicking on page:",btns[i].text) #page 1 button
    btns[i].click()
```

```python
# method5: 
# if you want to move to next page e.g# 2,3.
# just click on next button it will take you to next page
for i in range(4):
    driver.find_element("xpath", "//button[@class= 'btn btn-default next']").click()
    atv_btn = driver.find_element("xpath","//button[@class = 'btn btn-default page-link page active btn-primary']")
    print("clicking on page:",atv_btn.text)
    time.sleep(1)

    
```
